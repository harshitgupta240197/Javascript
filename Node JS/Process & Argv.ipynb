# `process` and `process.argv` in Node.js

In Node.js, `process` is a globally available object providing control and information about the currently running instance, while `process.argv` is a built-in property used to capture command-line arguments.

## The `process` Object

Because `process` is a global object, you do not need to use `require()` or `import` to use it. It acts as a bridge between your JavaScript code and the operating system.

### Key Properties & Methods

- **`process.env`** — Exposes environment variables (e.g. `process.env.NODE_ENV`).
- **`process.cwd()`** — Returns the current working directory.
- **`process.exit()`** — Exits the active program via a status code (e.g. `0` for success).
- **`process.pid`** — Returns the unique process ID assigned by the operating system.

## Understanding `process.argv`

The `argv` property stands for **argument vector**. It returns an array of strings containing everything passed to the terminal when executing the script.

The structure of the `process.argv` array always follows this exact sequence:

| Index | Content Description | Example Value |
| --- | --- | --- |
| `[0]` | Absolute file path of the Node.js executable | `'/usr/local/bin/node'` |
| `[1]` | Absolute file path of the JavaScript file being executed | `'/Users/project/app.js'` |
| `[2]` | The first user-provided argument | `'hello'` |
| `[3]` | The second user-provided argument | `'world'` |

## Code Example: Accessing User Arguments

To collect only the inputs provided by the user, use JavaScript's `.slice(2)` method to cut off the system paths found at indices 0 and 1.

Create a file named `app.js`:

```javascript
// app.js

// 1. View the entire array
console.log("Full process.argv array:", process.argv);

// 2. Extract only user inputs
const userArgs = process.argv.slice(2);
console.log("Cleaned user arguments:", userArgs);

// 3. Access arguments directly by index
const firstInput = userArgs[0];
console.log(`First custom input: ${firstInput}`);
```

If you run this script in your terminal:

```bash
node app.js admin --port=8080
```

The output will look like this:

```text
Full process.argv array: [
  '/usr/local/bin/node',
  '/Users/project/app.js',
  'admin',
  '--port=8080'
]
Cleaned user arguments: [ 'admin', '--port=8080' ]
First custom input: admin
```

## Robust Parsing with Libraries

Raw elements inside `process.argv` are always plain strings. If you pass complex flags (like `--port=8080` or `-v`), you have to manually parse them using string splits.

For production applications, developers typically use third-party libraries available on the npm registry to parse arguments into clean JavaScript objects:

- **yargs** — Excellent for robust commands, automatic help menus, and required flags.
- **commander** — The most widely used solution for complex command-line interfaces (CLIs).
- **minimist** — A lightweight option purely focused on parsing argument options.
