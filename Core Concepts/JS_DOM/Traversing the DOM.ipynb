# Traversing the DOM

In JavaScript, you can traverse the Document Object Model (DOM) by targeting either **Element** nodes (HTML tags only) or **Node** objects (which include tags, text whitespace, and comments). For most web development use cases, you will want to target Elements to avoid accidentally selecting empty spacing or line breaks in your HTML.

## Node vs. Element Properties

| Target Relation | Element-Only Property (recommended) | Any Node Property (includes text/comments) |
| --- | --- | --- |
| Immediate parent | `parentElement` | `parentNode` |
| All children | `children` | `childNodes` |
| First child | `firstElementChild` | `firstChild` |
| Last child | `lastElementChild` | `lastChild` |
| Next sibling | `nextElementSibling` | `nextSibling` |
| Previous sibling | `previousElementSibling` | `previousSibling` |

## 1. Traversing Upwards (Parents)

Moving up the tree allows you to find an element's container or a distant ancestor.

- `parentElement` — returns the immediate HTML element parent.
- `closest(selector)` — searches up the DOM tree for the nearest ancestor matching a specific CSS selector, returning `null` if none match.

```javascript
const child = document.querySelector('.my-child-item');

// Get the direct parent element
const directParent = child.parentElement;

// Find a specific ancestor further up the tree
const mainContainer = child.closest('.main-wrapper');
```

## 2. Traversing Downwards (Children)

Moving down the tree targets items nested inside a specific container.

- `children` — returns a live `HTMLCollection` containing only the child HTML elements.
- `firstElementChild` / `lastElementChild` — grabs the very first or last child tag.

```javascript
const parent = document.querySelector('.gallery');

// Get a list of all child tags
const allChildren = parent.children;

// Get specific positions
const firstCard = parent.firstElementChild;
const lastCard = parent.lastElementChild;

// Loop through elements (convert HTMLCollection to an Array first)
Array.from(parent.children).forEach(child => {
  child.classList.add('active');
});
```

## 3. Traversing Sideways (Siblings)

Moving sideways targets items that share the exact same parent node.

- `nextElementSibling` — selects the next adjacent HTML element on the same level.
- `previousElementSibling` — selects the preceding adjacent HTML element on the same level.

```javascript
const currentItem = document.querySelector('.step-2');

// Move to the next element
const nextStep = currentItem.nextElementSibling;

// Move to the previous element
const previousStep = currentItem.previousElementSibling;
```

## 4. Fetching All Siblings

JavaScript does not have a native `siblings()` method like jQuery. To gather all siblings, you must fetch the parent's children and filter out the original element:

```javascript
const element = document.querySelector('.target');

const siblings = Array.from(element.parentElement.children).filter(child => {
  return child !== element;
});
```
