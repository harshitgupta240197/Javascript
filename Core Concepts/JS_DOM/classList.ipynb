# element.classList

In JavaScript, `element.classList` is a read-only property that returns a live `DOMTokenList` collection of the class attributes of an HTML element. It provides a modern, convenient alternative to `element.className`, eliminating the need for complex string manipulation when managing CSS classes.

## Common Methods

| Method | What it does |
| --- | --- |
| `add(class1, class2, ...)` | Inserts one or more classes on the element. Ignores classes that already exist. |
| `remove(class1, class2, ...)` | Removes one or more classes from the element. |
| `toggle(className, force)` | Removes the class if it exists, or adds it if it does not. The optional second boolean parameter forces the addition (`true`) or removal (`false`). |
| `contains(className)` | Checks whether the element has the specified class. Returns `true` or `false`. |
| `replace(oldClass, newClass)` | Substitutes an existing class with a new one. Returns `true` if the replacement succeeded. |

## Code Example

```javascript
// Select the target HTML element
const box = document.querySelector('.box');

// 1. Add a single class or multiple classes
box.classList.add('active');
box.classList.add('highlight', 'rounded-corners');

// 2. Remove a class
box.classList.remove('highlight');

// 3. Toggle a class (adds if missing, removes if present)
box.classList.toggle('visible');

// 4. Check if a class exists
if (box.classList.contains('active')) {
  console.log('The box is active!');
}

// 5. Replace an old class with a new one
box.classList.replace('active', 'disabled');
```

## `classList` vs `className`

| Feature | `element.classList` | `element.className` |
| --- | --- | --- |
| Data type | `DOMTokenList` (object-like) | String |
| Handling multiple classes | Simple method arguments — `.add('a', 'b')` | Requires string concatenation — `+= ' a b'` |
| Checking for a class | Built-in `.contains()` method | Requires `.indexOf()` or a regex |
| Safety | Modifies only the specified target classes | Can accidentally overwrite the whole class string |
