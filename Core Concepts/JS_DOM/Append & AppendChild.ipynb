# `append()` vs `appendChild()`

The primary difference is that `append()` is a modern, flexible method that accepts both strings and multiple nodes, while `appendChild()` is an older, stricter DOM method that only accepts a single `Node` object.

## Quick Comparison

| Feature | `Element.append()` | `Node.appendChild()` |
| --- | --- | --- |
| Accepts strings? | Yes — converted to text nodes | No — throws a `TypeError` |
| Accepts multiple items? | Yes — comma-separated list | No — only one at a time |
| Return value | `undefined` | Returns the appended node |
| Interface | `Element` interface | `Node` interface |

## 1. Appending Text/Strings

`append()` handles plain text seamlessly by automatically converting it into a text node.

```javascript
const div = document.createElement('div');
div.append("Hello World"); // Works perfectly
```

`appendChild()` fails when passing plain text. You must explicitly create a text node first.

```javascript
const div = document.createElement('div');

// div.appendChild("Hello World"); // Error!

const textNode = document.createTextNode("Hello World");
div.appendChild(textNode); // Works, but requires extra code
```

## 2. Appending Multiple Items

`append()` allows you to pass multiple arguments to insert everything at once.

```javascript
const parent = document.createElement('div');
const p = document.createElement('p');
const span = document.createElement('span');

parent.append(p, span, "Some text"); // Appends all three sequentially
```

`appendChild()` only accepts one argument. If you pass multiple, it only appends the first one and ignores the rest.

```javascript
parent.appendChild(p, span); // Only appends <p>, ignores <span>
```

## 3. Return Values

- `append()` does not return anything (`undefined`).
- `appendChild()` returns the element that was just appended, which is useful if you need to keep a reference to it or modify it immediately.

```javascript
const parent = document.createElement('div');
const child = document.createElement('p');

const result = parent.appendChild(child);
console.log(result); // Logs: <p></p>
```
