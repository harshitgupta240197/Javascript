# Event Delegation

Event delegation is a pattern where you attach a **single** event listener to a parent element instead of adding individual listeners to many child elements.

It relies on [[event-bubbling|event bubbling]]: when an event fires on a child, it propagates up the DOM tree to its ancestors. By listening at a higher level, the parent intercepts the event and handles the logic dynamically.

## Why Use It

- **Memory efficiency** — hundreds of separate listeners create real overhead. One listener on the parent saves resources.
- **Dynamic content** — child elements injected into the DOM later inherit the parent's listener automatically, with no re-initialization.
- **Cleaner maintenance** — consolidating event logic reduces boilerplate and keeps DOM structure separate from interaction logic.

## Implementation

Use `event.target` to identify which child triggered the action, and `Element.closest()` to reliably locate the matching element even when the click lands on nested markup (like a `<span>` inside a button).

```js
// Grab the parent container element
const menuList = document.querySelector('#menu-list');

// Attach a single listener to the parent
menuList.addEventListener('click', (event) => {
  // Find the nearest matching ancestor
  const listItem = event.target.closest('.menu-item');

  // Guard clause: exit if the click happened outside a menu item
  if (!listItem) return;

  // Safely execute logic using the verified child element
  console.log(`User selected item ID: ${listItem.dataset.id}`);
});
```

## Key Mechanisms

| Mechanism | Description |
|---|---|
| `event.target` | The exact, deepest element where the event originated (e.g. the clicked button's text node's parent). |
| `event.currentTarget` | The element the handler is physically attached to (the parent container). |
| `event.target.closest()` | Traverses up from the target to find a matching ancestor — ensures inner tags don't break detection. |

## Limitations

- **Non-bubbling events** — delegation fails on native events that don't bubble: `focus`, `blur`, `scroll`, `mouseenter`, `mouseleave`. Use the bubbling equivalents (`focusin`, `focusout`, `mouseover`, `mouseout`) when you need delegation.
- **CPU overhead** — if the parent sits high in a busy DOM tree, running selector checks on every click adds a minor processing cost.
