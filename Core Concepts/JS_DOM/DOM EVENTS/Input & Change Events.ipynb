# `input` vs `change` Events

The core difference is **when they fire**: `input` triggers immediately every time the value changes, while `change` triggers only when the change is committed or the element loses focus.

## Direct Comparison

| Feature | `input` | `change` |
|---|---|---|
| **Timing for text fields** | Fires instantly on every keystroke, deletion, or paste. | Fires only when the user blurs the field or presses Enter. |
| **Timing for toggles/menus** | Fires immediately on click/select. | Fires immediately on click/select. |
| **Best used for** | Real-time features — live search, character counters, password strength meters. | Expensive operations — form validation, saving draft data to a database. |

## Code Examples

### 1. `input` — real-time tracking

Use when you need live updates as the user types.

```js
const textInput = document.querySelector('#username');

// Triggers instantly with every character typed or deleted
textInput.addEventListener('input', (event) => {
    console.log('Live value:', event.target.value);
});
```

### 2. `change` — finalized tracking

Use when you want to wait until the user finishes interacting with the field.

```js
const textInput = document.querySelector('#username');

// Triggers only when focus moves away from the input
textInput.addEventListener('change', (event) => {
    console.log('Committed value:', event.target.value);
});
```

## Behavior With Other Element Types

Text fields are the outlier. For controls where typing isn't involved, both events fire at virtually the same moment:

- **Checkboxes & radio buttons** — both fire the instant the option is checked or unchecked.
- **Dropdowns (`<select>`)** — both fire immediately when a new option is chosen.
- **Color / date / range pickers** — both fire as the user drags a slider or picks a value.

> [!note]
> Neither event fires when you change a value programmatically (e.g. `element.value = "new text"`). You must dispatch the event manually if you want listeners to run:
> ```js
> element.value = 'new text';
> element.dispatchEvent(new Event('input', { bubbles: true }));
> ```
