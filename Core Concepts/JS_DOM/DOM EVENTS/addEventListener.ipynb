# `addEventListener()`

The `addEventListener()` method attaches an event handler to a specified DOM element without overwriting existing event handlers. It allows JavaScript to "watch" for specific user interactions (like clicks, keypresses, or scrolls) and execute a function when that interaction happens.

## Basic Syntax

```javascript
element.addEventListener(event, function, options);
```

### Parameter Breakdown

- **`event`** — a required string specifying the name of the event (e.g. `'click'`, `'keydown'`, `'submit'`). Do not use the `"on"` prefix: use `'click'`, not `'onclick'`.
- **`function`** — a required callback function that runs when the event triggers. It automatically receives an `event` object containing details about the action.
- **`options`** *(optional)* — an object or boolean that customizes the listener behavior (e.g. controlling event propagation or executing the listener only once).

## Code Examples

### 1. Using an Anonymous Arrow Function

This is the most common approach for quick, inline operations.

```javascript
const button = document.getElementById('myButton');

button.addEventListener('click', () => {
  alert('Button was clicked!');
});
```

### 2. Using a Named Function

Using a named function keeps code organized and allows you to remove the listener later if needed.

```javascript
const button = document.getElementById('myButton');

function handleInteraction(event) {
  console.log('Event type:', event.type); // Outputs: "click"
}

// Add the listener
button.addEventListener('click', handleInteraction);
```

### 3. Utilizing the Optional Configuration Object

You can pass an options object as the third argument to unlock finer controls:

```javascript
button.addEventListener('click', () => {
  console.log('This will only run one time.');
}, { once: true }); // Automatically deletes itself after firing once
```

## Core Advantages over `element.onclick`

- **Multiple handlers** — you can attach multiple independent functions to the exact same event on a single element. `onclick` will overwrite previous handlers.
- **Clean separation** — keeps your behavior logic completely out of your structural HTML files.
- **Removal capability** — easily clean up memory and performance using the matching `removeEventListener()` method.
