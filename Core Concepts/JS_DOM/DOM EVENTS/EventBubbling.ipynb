# Event Bubbling

Event bubbling is the mechanism where an event triggered on a nested child element propagates **upward** through its ancestors in the DOM tree.

Click a button inside a `<div>` and the `click` event fires on the button first, then on the `<div>`, then on `<body>`, and continues up to `document` and `window`. This upward flow is the default behavior for most browser events.

## The Three Phases of Event Propagation

Per the W3C DOM spec, events travel in three distinct phases:

| Phase | Direction | Description |
|---|---|---|
| **1. Capturing** | Downward | The event travels from `window` down through the target's ancestors. |
| **2. Target** | At target | The event reaches the element that actually triggered the interaction. |
| **3. Bubbling** | Upward | The event travels back up from the target to `window`. |

## How Bubbling Works

Consider this nested structure:

```html
<div id="parent" style="padding: 20px; background: lightgray;">
  Parent DIV
  <button id="child">Click Me!</button>
</div>
```

Attach listeners to both elements and they execute from the inside out:

```js
document.getElementById('parent').addEventListener('click', () => {
  console.log('Parent handler triggered');
});

document.getElementById('child').addEventListener('click', () => {
  console.log('Child handler triggered');
});

// Clicking the button logs:
// 1. "Child handler triggered"
// 2. "Parent handler triggered"
```

## Controlling Bubbling

### 1. Stopping propagation

To prevent an event from rising to parent elements, call `event.stopPropagation()` inside the handler:

```js
document.getElementById('child').addEventListener('click', (event) => {
  event.stopPropagation(); // Stops the event from bubbling up to #parent
  console.log('Child clicked. Bubbling stopped.');
});
```

### 2. `target` vs. `currentTarget`

When handling bubbled events, two properties tell you where the event came from:

- **`event.target`** — the deepest, original element that triggered the event (the button).
- **`event.currentTarget`** — the element currently running the handler (the parent `<div>`).

## Practical Application: Event Delegation

Bubbling enables an efficient pattern called **event delegation**. Instead of attaching listeners to dozens of child elements, attach a single listener to the parent:

```js
// A single listener handles clicks for all current and future list items
document.getElementById('myList').addEventListener('click', (event) => {
  if (event.target.tagName === 'LI') {
    console.log(`Clicked item text: ${event.target.textContent}`);
  }
});
```

**Why it wins:**

- **Fewer listeners** — saves memory.
- **Dynamic elements** — automatically covers elements added to the DOM later.
