# The `onclick` Property

The `onclick` property in JavaScript is a DOM property used to execute a specific piece of JavaScript code or trigger a function when a user clicks on an HTML element.

## Ways to Implement `onclick`

You can handle click interactions using three primary methods.

### 1. Inline HTML Attribute

You can assign code directly inside your HTML tag. This approach is easy for small scripts but is generally discouraged for larger projects because it mixes HTML and JavaScript logic.

```html
<button onclick="alert('Button clicked!')">Click Me</button>
```

### 2. JavaScript DOM Property

You can target the element in your JavaScript file and assign a function directly to its `onclick` property.

> **Crucial note:** When assigning a function name, do not include parentheses `()`, or it will execute immediately when the page loads.

```javascript
const myButton = document.getElementById("btn");

// Assign the function reference directly
myButton.onclick = function() {
    console.log("Element clicked!");
};
```

### 3. Modern Approach: `addEventListener`

While `onclick` is fast, using `addEventListener("click", ...)` is considered best practice. Unlike `onclick` — which can only hold one function at a time and overwrites previous assignments — `addEventListener` allows you to attach multiple functions to the same element.

```javascript
const myButton = document.getElementById("btn");

myButton.addEventListener("click", () => {
    console.log("First listener executed!");
});
```

## Direct Comparison: `onclick` vs `addEventListener`

| Feature | `onclick` property | `addEventListener("click")` |
| --- | --- | --- |
| Max listeners | Only one function can be assigned | Multiple functions can be attached |
| Overwriting | Overwrites any existing click functions | Appends safely without removing old actions |
| Code splitting | Couples behavior directly to the element property | Promotes cleaner, separated script files |

## Accessing the Event Object and `this`

Inside an `onclick` function handler, you can pass the `event` parameter to gain control over the interaction details (such as preventing the default link navigation via `event.preventDefault()`). Additionally, the keyword `this` refers directly to the HTML element that fired the click event.

```javascript
myButton.onclick = function(event) {
    console.log(event.type); // Outputs: "click"
    this.style.backgroundColor = "blue"; // Turns this button blue
};
```
