# JavaScript Events

JavaScript events are signals fired by the browser window or DOM elements to notify your code that a specific interaction or system change has occurred. They enable web pages to react dynamically to user input, such as clicks, key presses, or form submissions.

## Common Event Types

Events are grouped into categories based on what triggers them:

- **Mouse events** — triggered by pointer interactions like `click`, `dblclick`, `mousemove`, `mouseover`, and `mouseout`.
- **Keyboard events** — fired when interacting with keys, including `keydown` and `keyup`.
- **Form events** — related to input controls, such as `submit`, `change`, `focus`, `blur`, and `input`.
- **Window/document events** — fired by the browser lifecycle, like `load`, `resize`, `scroll`, and `DOMContentLoaded`.

## How to Listen for Events

While you can attach events inline inside HTML attributes (e.g. `onclick="..."`), using the `addEventListener()` method in your JavaScript file is the best practice. It separates your logic from the structure and allows multiple listeners on a single element.

```javascript
// Select the element
const button = document.querySelector('#myButton');

// Add the event listener
button.addEventListener('click', (event) => {
  console.log('Button was clicked!');
  console.log('Target element:', event.target);
});
```

## Key Event Concepts

- **The event object (`e`)** — the browser automatically passes an object containing details about the event (like `event.target`, cursor coordinates, or the pressed key) to the callback function.
- **Event propagation** — events propagate through the DOM in two phases: *capturing* (traveling down from the root to the target) and *bubbling* (traveling back up from the target to the root).
- **Prevent default** — `event.preventDefault()` stops the browser's default behavior, such as preventing a form from refreshing the page on submission.
- **Stop propagation** — `event.stopPropagation()` stops the event from bubbling up to parent elements.
