# Keyboard Events in JavaScript

Keyboard events in JavaScript allow you to track and respond to user interactions with a physical keyboard. Whenever a user interacts with a key, the browser triggers a specific event and generates a `KeyboardEvent` object containing precise information about that action.

## 1. Types of Keyboard Events

There are two primary modern keyboard events:

- **`keydown`** — Fires the moment a key is pressed down. It repeats continuously if the key is held down.
- **`keyup`** — Fires when the user releases the key.

> [!warning] Note on `keypress`
> You may encounter the `keypress` event in legacy code, but it is **deprecated and obsolete**. Do not use it for modern applications.

## 2. The KeyboardEvent Object

When a keyboard event occurs, an event object (usually named `event` or `e`) is automatically passed to your callback function. This object contains several crucial properties.

### Essential Properties

- **`event.key`** — Returns the character value represented by the key (e.g. `"a"`, `"A"`, `"Enter"`, `"ArrowUp"`). This property is case-sensitive and adjusts based on the active keyboard layout or whether the Shift key is active.
- **`event.code`** — Returns the physical location of the key on a standard QWERTY keyboard (e.g. `"KeyA"`, `"Enter"`, `"ArrowUp"`). This value never changes, regardless of your keyboard layout or language setting.
- **`event.repeat`** — A boolean that returns `true` if the key is being held down to trigger repeated `keydown` events.

### Modifier Key Properties (Booleans)

These properties return `true` if the respective modifier key was held down while pressing the primary key:

| Property | Key |
| --- | --- |
| `event.ctrlKey` | Control |
| `event.shiftKey` | Shift |
| `event.altKey` | Alt / Option |
| `event.metaKey` | Windows / Command |

## 3. Quick Comparison: `event.key` vs `event.code`

| User Action | `event.key` Value | `event.code` Value |
| --- | --- | --- |
| Pressing the `a` key | `"a"` | `"KeyA"` |
| Pressing `Shift` + `a` | `"A"` | `"KeyA"` |
| Pressing `Spacebar` | `" "` | `"Space"` |

## 4. Implementation Code Example

To listen for keyboard events, you typically attach an event listener using `addEventListener()` to the entire document or to a specific input field:

```javascript
// Listen to any key pressed on the page
document.addEventListener('keydown', (event) => {
  console.log(`Key pressed: ${event.key}`);
  console.log(`Physical code: ${event.code}`);

  // Example: Detect a keyboard shortcut (Ctrl + S)
  if (event.ctrlKey && event.key === 's') {
    event.preventDefault(); // Stop the browser's default "Save Page" window
    console.log('Custom save functionality triggered!');
  }

  // Example: Detect specific action keys
  if (event.key === 'ArrowUp') {
    console.log('Player moved up!');
  }
});
```
