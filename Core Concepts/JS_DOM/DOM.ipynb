# DOM (Document Object Model)

The **Document Object Model (DOM)** is a programming interface provided by the browser that represents a web page as a hierarchical tree of objects. It acts as a bridge that allows JavaScript to dynamically access, modify, and update the structure, content, and styling of an HTML document without reloading the page.

When a browser loads an HTML page, it parses the code and creates this object tree, where every HTML element, attribute, and piece of text is a separate **node**.

---

## 1. Selecting Elements (Querying)

Before you can change anything on a webpage, you must tell JavaScript which element to target using methods provided by the [Document interface](https://developer.mozilla.org/en-US/docs/Web/API/Document_Object_Model).

| Method | Returns |
| --- | --- |
| `document.getElementById('id')` | A single element by its unique ID |
| `document.getElementsByClassName('class')` | All elements with a specific class |
| `document.querySelector('selector')` | The **first** element matching a CSS selector |
| `document.querySelectorAll('selector')` | **All** elements matching a CSS selector |

```javascript
// Example: Targeting a specific paragraph
const mainHeading = document.getElementById('main-title');
const buttons = document.querySelectorAll('.btn-submit');
```

---

## 2. Modifying Content & Styles

Once selected, you can alter the text, HTML contents, or inline CSS styles of an element using properties like `.textContent`, `.innerHTML`, or `.style`.

```javascript
// Changing text content safely
mainHeading.textContent = "Welcome to My Updated Website!";

// Inserting HTML content
const container = document.querySelector('.container');
container.innerHTML = "<p>New paragraph nested inside the container.</p>";

// Altering styles (uses camelCase instead of hyphenated CSS names)
mainHeading.style.color = "blue";
mainHeading.style.fontSize = "24px";
```

> [!note] `textContent` vs `innerHTML`
> `textContent` treats the input as plain text. `innerHTML` parses it as markup, so avoid it with untrusted input.

---

## 3. Creating & Removing Elements

You can dynamically build new pieces of a website or remove components entirely using structural [Node interface](https://developer.mozilla.org/en-US/docs/Web/API/Node) methods.

```javascript
// 1. Create a new element
const newBox = document.createElement('div');

// 2. Configure the element
newBox.textContent = "I am a new box!";
newBox.className = "box-style";

// 3. Append it to an existing element in the DOM tree
document.body.appendChild(newBox);

// 4. Removing an element
newBox.remove();
```

---

## 4. Handling User Events

The DOM lets you listen for user actions (clicks, keypresses, mouse movements) and trigger JavaScript in response.

```javascript
const actionButton = document.querySelector('#action-btn');

// Trigger code when the user clicks the button
actionButton.addEventListener('click', function() {
    alert('Button was clicked!');
});
```
