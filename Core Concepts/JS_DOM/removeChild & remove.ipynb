# `remove()` vs `removeChild()`

The primary difference between `remove()` and `removeChild()` is how they target the element and what they return. `element.remove()` is called directly on the element you want to delete and returns `undefined`, whereas `parentNode.removeChild(childNode)` is called on the parent of the target element and returns the removed node so it can be reused.

## Summary of Differences

| Feature | `remove()` | `removeChild()` |
| --- | --- | --- |
| Called on | The target element itself | The parent of the target element |
| Syntax | `element.remove()` | `parent.removeChild(child)` |
| Return value | `undefined` | The removed node object |
| Memory behavior | Deletes element permanently (if unreferenced) | Holds element in memory for potential relocation |

## 1. `Element.remove()`

This is the modern and simpler approach. You select the precise element you want to get rid of and eliminate it directly from the DOM tree.

```javascript
// Select the element directly
const item = document.getElementById("bad-apple");

// Remove it directly without needing its parent
item.remove();
```

## 2. `Node.removeChild()`

This is an older, legacy method from DOM Level 1. It requires you to have a reference to the parent element first. Because it returns the removed element, it is highly useful if you want to pull an item out of one list and immediately move it to another.

```javascript
const list = document.getElementById("shopping-list");
const item = document.getElementById("milk");

// Pass the target element as an argument to the parent
const removedItem = list.removeChild(item);

// The element is gone from 'list', but saved in 'removedItem'
document.getElementById("cart").appendChild(removedItem);
```

If you do not have direct access to the parent node but still want to use `removeChild()`, you must access the parent dynamically via the `.parentNode` property:

```javascript
const item = document.getElementById("milk");
item.parentNode.removeChild(item);
```
