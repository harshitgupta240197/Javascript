# HTML Attributes in JavaScript

**Attributes** are the initial key-value pairs defined within HTML tags — `id="main"`, `src="image.png"`, `class="active"`. When the browser parses the HTML document, it builds the DOM, where these attributes can be interacted with using specific JavaScript methods.

---

## 🛠️ Core Methods for Managing Attributes

Four main built-in methods on DOM elements:

| Method | Purpose |
| --- | --- |
| `getAttribute(name)` | Fetches the current text value of a specified attribute |
| `setAttribute(name, value)` | Changes the value of an existing attribute, or adds it if absent |
| `removeAttribute(name)` | Deletes the specified attribute completely from the element |
| `hasAttribute(name)` | Checks if the attribute exists, returning a boolean |

```javascript
const link = document.querySelector('a');

// 1. Set an attribute
link.setAttribute('href', 'https://example.com');

// 2. Get an attribute value
const url = link.getAttribute('href'); // Returns 'https://example.com'

// 3. Check for presence
if (link.hasAttribute('target')) {
  // 4. Remove an attribute
  link.removeAttribute('target');
}
```

## ⚠️ Attributes vs. Properties

While they seem identical, JavaScript draws a clear line between **HTML attributes** and **DOM properties**:

| Feature | HTML Attributes | DOM Properties |
| --- | --- | --- |
| **Definition** | Written directly in the HTML markup | Variables attached to JavaScript DOM objects |
| **Data type** | Always returns a **string** | Can be strings, booleans, objects, etc. |
| **State** | Represents the **initial/default** state | Represents the **live/current** state |
| **Case sensitivity** | Case-insensitive | Case-sensitive |

### The value/input mismatch

A classic example of this difference is a text `<input>` field:

```javascript
const input = document.querySelector('input'); // HTML: <input value="Hello">

// If a user types "World" into the box:
console.log(input.getAttribute('value')); // Still outputs "Hello" (original HTML state)
console.log(input.value);                 // Outputs "World" (current user property state)
```

### Naming anomalies

Because `class` is a reserved word in JavaScript, the HTML `class` attribute must be accessed as the `className` or `classList` property in DOM scripting:

```javascript
const div = document.querySelector('div');

div.setAttribute('class', 'main-box'); // Attribute syntax
console.log(div.className);            // Property syntax ("main-box")
```

## 🏷️ Working with Custom `data-*` Attributes

HTML allows developers to embed custom data fields using the `data-` prefix. In JavaScript, these are automatically gathered into a built-in property object called `dataset`.

> [!note] Naming conversion
> Dashes in the HTML attribute name become **camelCase** keys inside the `dataset` object — `data-user-id` → `dataset.userId`.

```html
<!-- HTML Structure -->
<div id="user" data-user-id="9876" data-role="admin"></div>
```

```javascript
const userDiv = document.getElementById('user');

// Accessing via dataset (recommended)
console.log(userDiv.dataset.userId); // Outputs: "9876" (camelCase conversion)
console.log(userDiv.dataset.role);   // Outputs: "admin"

// Updating values
userDiv.dataset.role = 'editor'; // Updates HTML to data-role="editor"
```
