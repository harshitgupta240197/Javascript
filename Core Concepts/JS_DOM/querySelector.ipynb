# document.querySelector()

The `document.querySelector()` method returns the **first** HTML element that matches a specified CSS selector within the document. If no matches are found, it returns `null`.

It is one of the most flexible and widely used methods for DOM manipulation because it accepts any valid CSS selector string.

---

## 💡 Syntax

```javascript
const element = document.querySelector('css-selector');
```

---

## 🔍 Common Target Selectors

**By tag name** — targets the first element of that HTML tag.

```javascript
const firstParagraph = document.querySelector('p'); // <p>
```

**By class name** — requires a period (`.`) prefix.

```javascript
const activeItem = document.querySelector('.active'); // class="active"
```

**By ID name** — requires a hashtag (`#`) prefix.

```javascript
const mainHeader = document.querySelector('#header'); // id="header"
```

**By attribute** — uses square brackets `[]`.

```javascript
const emailInput = document.querySelector('input[type="email"]');
```

**By complex / nested selectors** — combines selectors just like in CSS rules.

```javascript
const sidebarLink = document.querySelector('.sidebar ul li a');
```

---

## 🔄 querySelector vs querySelectorAll

When working with multiple elements, you must choose between getting a single element or a list:

| Feature | `querySelector()` | `querySelectorAll()` |
| --- | --- | --- |
| **Return value** | The first matching element object | A static `NodeList` of all matches |
| **No match found** | Returns `null` | Returns an empty `NodeList` |
| **Direct manipulation** | Yes — e.g. `el.style.color = 'red'` | No — must loop via `.forEach()` first |

---

## 🛠️ Practical Example

Consider the following HTML block:

```html
<ul id="menu">
  <li class="item">Home</li>
  <li class="item">About</li>
</ul>
```

You can target and manipulate the elements like this:

```javascript
// 1. Select the first item only ("Home")
const firstItem = document.querySelector('.item');
firstItem.style.color = 'blue';

// 2. Scoped selection (searching inside a specific parent element)
const menu = document.querySelector('#menu');
const nestedItem = menu.querySelector('.item'); // Faster and scoped
```

> [!tip] Scoped queries
> `querySelector()` isn't limited to `document` — any element can call it. Scoping the search to a parent narrows the tree being traversed and avoids accidental matches elsewhere on the page.
