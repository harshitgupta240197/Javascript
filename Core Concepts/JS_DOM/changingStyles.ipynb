# Changing Styles in the DOM

To change styles in the Document Object Model (DOM) using JavaScript, you can manipulate inline styles, manage CSS classes, or update CSS custom properties (variables). The choice of method depends on whether you are making a quick layout tweak or handling complex application styling.

## Method 1: The `.style` Property

*Best for a single inline tweak.*

Every DOM element has a `style` object that lets you modify inline properties directly.

- **CamelCase transformation:** Standard hyphenated CSS properties (like `background-color`) must be written in camelCase (`backgroundColor`), because the dash character represents subtraction in JavaScript.
- **Value format:** All values must be assigned as a string and include units (e.g. `'20px'`, `'10rem'`, `'50%'`).

```javascript
const element = document.querySelector('.my-element');

// Update individual inline properties
element.style.backgroundColor = 'royalblue';
element.style.fontSize = '24px';
element.style.marginTop = '2rem';
```

### Overwriting multiple styles

If you need to apply an entire raw string of CSS text at once, use the `.style.cssText` property.

> **Note:** This completely overwrites any pre-existing inline styles on that element.

```javascript
element.style.cssText = 'color: white; border: 1px solid red; padding: 10px;';
```

## Method 2: The `.classList` Object

*Best practice for production.*

Instead of hardcoding styles inside your JavaScript files, it is cleaner to define styles inside an external CSS stylesheet and use JavaScript to add, remove, or cycle through class names. This technique decouples design rules from behavioral code.

```javascript
const element = document.querySelector('.card');

// 1. Add a single or multiple CSS classes
element.classList.add('active', 'highlighted');

// 2. Remove specific classes
element.classList.remove('hidden');

// 3. Toggle a class (adds if missing, removes if present)
element.classList.toggle('dark-mode');

// 4. Check if a class is currently applied (returns true/false)
if (element.classList.contains('active')) {
  console.log('Element is active!');
}
```

## Method 3: Manipulating CSS Variables

*Best for global themes.*

If you want to modify a shared theme layout property globally, you can dynamically adjust a CSS custom property (variable) via the global document style blueprint.

```css
/* CSS definition */
:root {
  --main-theme-color: #333333;
}
```

```javascript
// JavaScript to update the root variable instantly across the entire page
document.documentElement.style.setProperty('--main-theme-color', '#ff5733');
```

## Summary Checklist

| Objective | Recommended Method | Code Example |
| --- | --- | --- |
| Quick dynamic calculation (e.g. following a mouse cursor tracking position) | `element.style.propertyName` | `element.style.left = event.clientX + 'px';` |
| Toggling global page behaviors (e.g. showing a modal, dark mode transition) | `element.classList` methods | `element.classList.toggle('open');` |
| Overwriting all styling properties completely | `element.style.cssText` | `element.style.cssText = 'display: none;';` |
| Mutating wide-scale structural tokens dynamically | `.style.setProperty()` | `root.style.setProperty('--brand-padding', '5px');` |
