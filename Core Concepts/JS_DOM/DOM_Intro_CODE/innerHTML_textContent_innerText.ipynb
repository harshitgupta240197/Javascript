# innerHTML vs textContent vs innerText

The primary difference between `innerHTML`, `textContent`, and `innerText` lies in **how they read and write content inside an HTML element**:

- `innerHTML` handles raw HTML tags
- `textContent` retrieves all text, including hidden content
- `innerText` retrieves only the visually rendered text, as seen by a human user

---

## 📊 Summary Comparison

| Feature | `innerHTML` | `textContent` | `innerText` |
| --- | --- | --- | --- |
| **Returns HTML tags?** | Yes | No | No |
| **Includes hidden text?** | Yes | Yes (even `display: none`) | No (obeys CSS) |
| **Performance** | Slowest (parses HTML) | Fastest (raw text retrieval) | Moderate (triggers layout reflow) |
| **Security risk** | High (susceptible to XSS) | Safe (escapes HTML tags) | Safe (escapes HTML tags) |

---

## Deep Dive into Each Property

### 1. `innerHTML`

**Behavior:** Gets or sets the markup contained within the element.

- **Reading** — returns the full string of raw text combined with any child HTML tags.
- **Writing** — parses strings containing HTML tags into working DOM elements.

> [!warning] Security
> Avoid passing unvalidated user input directly into `innerHTML` — it exposes your site to **Cross-Site Scripting (XSS)** attacks.

### 2. `textContent`

**Behavior:** Gets or sets the text content of a node and all of its descendants.

- **Reading** — extracts every piece of text inside the element, including text inside hidden nodes (such as `<script>`, `<style>`, or items hidden via CSS `display: none`). It returns the raw text exactly as it appears in the code, preserving line breaks and spaces.
- **Writing** — treats the input string strictly as plain text, converting characters like `<` or `>` into text literals rather than HTML elements.

### 3. `innerText`

**Behavior:** Gets or sets the visually rendered text content of a node.

- **Reading** — aims to match the text exactly as a human user would select it with a mouse cursor. It respects CSS, meaning it completely ignores text nested inside elements styled with `display: none` or `visibility: hidden`.

> [!note] Performance
> Because it relies on how things look visually, reading `innerText` forces the browser to calculate the page's layout (triggering a **reflow**), making it computationally heavier than `textContent`.

---

## 🧪 Code Example

Consider this HTML block:

```html
<div id="example">
  Hello <span style="display: none;">Secret</span> <b>World!</b>
</div>
```

Querying the element with all three properties:

```javascript
const element = document.getElementById("example");

console.log(element.innerHTML);
// Output: "  Hello <span style=\"display: none;\">Secret</span> <b>World!</b>\n"

console.log(element.textContent);
// Output: "  Hello Secret World!\n"  (includes hidden "Secret" and original spacing)

console.log(element.innerText);
// Output: "Hello World!"  (excludes hidden "Secret" and collapses extra spacing)
```

---

## ✅ Best Practices

1. **Use `textContent` by default** whenever you need to read or alter text inside an element — it is faster and more secure than the alternatives.
2. **Use `innerHTML` only** when you specifically intend to inject or build HTML tags dynamically, and ensure all data sources are fully sanitized.
3. **Use `innerText` only** when the current CSS styling of elements must be factored into the plain text data you are fetching.
