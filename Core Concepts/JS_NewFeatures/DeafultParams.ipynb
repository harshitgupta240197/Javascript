# Default Parameters in JavaScript

Default parameters allow you to initialize function parameters with default values if no value or `undefined` is passed to the function. Introduced in ES6, they eliminate the need to manually check for missing arguments inside the function body.

```javascript
function greet(name = "Guest") {
  return `Hello, ${name}!`;
}

console.log(greet());        // "Hello, Guest!" (uses default)
console.log(greet("Alice")); // "Hello, Alice!" (overrides default)
```

## Core Behavior & Rules

- **Triggered by `undefined` only** — The default is applied if an argument is missing or explicitly passed as `undefined`. Passing `null`, `false`, `0`, or `""` will *not* trigger the default, because those are valid values.
- **Left-to-right evaluation** — Parameters are evaluated sequentially. Later defaults can reference earlier parameters that have already been initialized.
- **Evaluated at call time** — The default expression runs anew on every call. If you assign an array or object as a default, a brand-new object is instantiated on each invocation.

## Key Usage Scenarios

### 1. Referencing Previous Parameters

A default parameter can use the value of a preceding parameter:

```javascript
function calculateTotal(price, tax = price * 0.1) {
  return price + tax;
}

console.log(calculateTotal(100)); // 110
```

### 2. Using Functions as Default Values

You can call a function to generate a dynamic default value:

```javascript
function generateId() {
  return Math.random().toString(36).substr(2, 9);
}

function createUser(name, id = generateId()) {
  return { name, id };
}
```

### 3. Default Parameters with Destructuring

Combine default parameters with object destructuring to safely handle optional configuration objects:

```javascript
function configureTheme({ color = "blue", width = 300 } = {}) {
  console.log(color, width);
}

configureTheme(); // blue 300
```

The trailing `= {}` falls back to an empty object when no argument is passed, so destructuring then pulls the per-property defaults instead of throwing.

## Best Practices

- **Place defaults at the end** — Put parameters with default values after required parameters. If a default parameter comes first, callers must explicitly pass `undefined` to skip it, which hurts readability.
- **Avoid scope conflicts** — You cannot reference variables or functions declared inside the function body from the default parameter initializer list. Defaults are evaluated in their own intermediate scope, which sits between the outer scope and the function body, so body-scoped bindings aren't visible yet and will throw a `ReferenceError`.
