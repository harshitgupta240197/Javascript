# Rest Parameters in JavaScript

The rest parameter syntax (`...`) allows a function to accept an indefinite number of arguments and bundle them into a single, **true array**. This makes it easy to handle functions where you don't know beforehand how many arguments will be passed.

```javascript
function sum(...numbers) {
  return numbers.reduce((total, num) => total + num, 0);
}

console.log(sum(1, 2));       // Output: 3
console.log(sum(1, 2, 3, 4)); // Output: 10
```

## 🛠️ Syntax Rules

- **Must be last** — the rest parameter must be the final parameter in the function declaration.
- **Only one allowed** — you cannot have multiple rest parameters in a single function signature.
- **No trailing commas** — you cannot place a comma after a rest parameter.
- **No default values** — rest parameters cannot have an initial default value assigned.

## 💡 Combining with Regular Parameters

Capture positional arguments first, then collect the leftovers with the rest parameter:

```javascript
function showTeam(captain, coCaptain, ...players) {
  console.log(`Captain: ${captain}`);
  console.log(`Co-Captain: ${coCaptain}`);
  console.log(`Players:`, players); // players is a true array
}

showTeam("Alex", "Sam", "Jordan", "Taylor", "Morgan");
// Captain: Alex
// Co-Captain: Sam
// Players: ['Jordan', 'Taylor', 'Morgan']
```

## ⚡ Rest Parameters vs. the `arguments` Object

Before ES6, developers used the built-in `arguments` object to handle dynamic parameters. Rest parameters are preferred today for several reasons:

| Feature | Rest Parameters (`...args`) | The `arguments` Object |
|---|---|---|
| **Type** | True `Array` instance | Array-like object (lacks array methods) |
| **Methods** | Supports `map`, `filter`, `reduce`, etc. | Requires conversion to use array methods |
| **Arrow functions** | Fully supported | Not available in arrow functions |
| **Scope** | Only captures un-named "leftover" arguments | Always captures every argument passed |

## 🔀 Rest vs. Spread

Both use the triple-dot (`...`) syntax but perform opposite operations:

- **Rest parameter** — gathers multiple individual arguments into a single array. Used in *function parameter lists*.
- **Spread operator** — expands a single array or iterable into individual separate values. Used in *function calls and array/object literals*.
