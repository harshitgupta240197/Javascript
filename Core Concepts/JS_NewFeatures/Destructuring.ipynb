# Destructuring in JavaScript

Destructuring is a concise syntax for unpacking values from arrays, properties from objects, and arguments from function parameters into distinct variables. It eliminates repetitive code and makes data extraction immediate and readable.

## Object Destructuring

Object destructuring extracts values by matching **property keys**. The order of the variables doesn't matter.

```javascript
const user = { id: 101, name: 'Alice', role: 'Admin' };

// Basic destructuring
const { name, role } = user;
console.log(name); // 'Alice'

// Renaming variables (aliasing)
const { name: userName, id: userId } = user;
console.log(userName); // 'Alice'

// Default values (fallback if property doesn't exist)
const { theme = 'dark' } = user;
console.log(theme); // 'dark'
```

## Array Destructuring

Array destructuring extracts values based on their **positional index**, using square brackets `[]`.

```javascript
const colors = ['red', 'green', 'blue'];

// Basic destructuring
const [firstColor, secondColor] = colors;
console.log(firstColor); // 'red'

// Skipping elements using commas
const [primary, , tertiary] = colors;
console.log(tertiary); // 'blue'

// Collecting remaining items with rest (...)
const [head, ...tail] = colors;
console.log(tail); // ['green', 'blue']
```

## Parameter Destructuring

You can destructure objects and arrays directly inside a function's parameter list. This is widely used in modern frameworks like React for handling configuration objects and component props.

### 1. Destructuring Object Parameters

```javascript
const product = { title: 'Laptop', price: 999, stock: 5 };

// Destructuring directly in the parameter list
function displayProduct({ title, price, discount = 0 }) {
  console.log(`${title} costs $${price}. Discount is ${discount}%`);
}

displayProduct(product); // 'Laptop costs $999. Discount is 0%'
```

### 2. Destructuring Array Parameters

```javascript
const coordinates = [40.7128, -74.0060];

function printLocation([lat, lng]) {
  console.log(`Latitude: ${lat}, Longitude: ${lng}`);
}

printLocation(coordinates); // 'Latitude: 40.7128, Longitude: -74.0060'
```

## Advanced: Arrays of Objects

Real-world data — API responses especially — often arrives as an array of objects. You can combine both syntax styles to pluck deep values in one expression.

```javascript
const users = [
  { id: 1, name: 'Alex' },
  { id: 2, name: 'Blake' }
];

// 1. Unpacking an object from a specific index
const [{ name: firstUserName }] = users;
console.log(firstUserName); // 'Alex'

// 2. Destructuring inside a loop (very common)
for (const { id, name } of users) {
  console.log(`User ID ${id} is ${name}`);
}
```

## Important Edge Cases

- **Null or undefined** — destructuring `null` or `undefined` throws a `TypeError`. Make sure the source data exists, or provide a default empty object/array (`function f({ a } = {}) { ... }`).
- **Rest syntax placement** — the rest operator (`...`) must always be the last element in a destructuring pattern.
