# Spread Operator — Operations Reference

The spread operator (`...`) expands or "unpacks" elements from an iterable (like an array or string) or properties from an object into individual values. It's primarily used to copy, merge, and pass data structures cleanly without mutating the original source.

## Array Operations

**Cloning arrays** — creates a shallow copy so modifications don't affect the original:

```javascript
const original = [1, 2, 3];
const copy = [...original]; // [1, 2, 3]
```

**Merging arrays** — combines multiple arrays into a new single array:

```javascript
const listA = [1, 2];
const listB = [3, 4];
const combined = [...listA, ...listB]; // [1, 2, 3, 4]
```

**Adding elements** — inserts elements at any position during initialization:

```javascript
const items = [2, 3];
const finalItems = [1, ...items, 4]; // [1, 2, 3, 4]
```

## Object Operations

**Cloning & overriding** — shallow-copies an object while updating or adding properties in the same expression:

```javascript
const user = { name: "Alice", role: "User" };
const admin = { ...user, role: "Admin" }; // { name: "Alice", role: "Admin" }
```

**Merging objects** — combines properties from separate objects. Later properties overwrite earlier ones when keys match:

```javascript
const target = { a: 1 };
const source = { b: 2, c: 3 };
const merged = { ...target, ...source }; // { a: 1, b: 2, c: 3 }
```

## Function Arguments

**Unpacking arguments** — converts an array into sequential arguments at the call site:

```javascript
const numbers = [5, 12, 8];
Math.max(...numbers); // Returns 12 (equivalent to Math.max(5, 12, 8))
```

## Important Distinctions

- **Shallow copy limitation** — spread only duplicates the top level. Nested objects or arrays keep their original references.
- **Spread vs. rest** — the syntax is identical (`...`) but the behavior is opposite. Spread unpacks a collection into individual elements; rest parameters gather multiple individual elements into a single array.
