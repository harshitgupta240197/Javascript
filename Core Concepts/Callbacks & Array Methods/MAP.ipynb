# `Array.prototype.map()` in JavaScript

The `Array.prototype.map()` method in JavaScript creates a new array populated with the results of calling a provided callback function on every single element in the original array. It is a **non-mutating** method, meaning it transforms data without altering the original array.

## The Syntax

```javascript
const newArray = originalArray.map(function callback(element, index, array) {
  // Return the transformed element
}, thisArg);
```

## The Callback Arguments

The callback function automatically receives three arguments from `map()` on every iteration:

1. **`element`** *(Required)* — The current item being processed in the array.
2. **`index`** *(Optional)* — The index number of the current item (starting from `0`).
3. **`array`** *(Optional)* — A reference to the original array being looped through.

## Code Examples

### 1. Basic Transformation (Arrow Function)

Using only the first argument (`element`) to double numbers.

```javascript
const numbers = [1, 2, 3, 4];
const doubled = numbers.map(num => num * 2);

console.log(doubled); // Output: [2, 4, 6, 8]
console.log(numbers); // Output: [1, 2, 3, 4] (Original is unchanged)
```

### 2. Extracting Object Properties

Commonly used in web applications to pull specific fields out of an array of API objects.

```javascript
const users = [
  { name: "Alice", age: 25 },
  { name: "Bob", age: 30 }
];

const names = users.map(user => user.name);
console.log(names); // Output: ["Alice", "Bob"]
```

### 3. Using the `index` Argument

Useful when you need the position of the element to modify the value.

```javascript
const items = ["Apple", "Banana", "Orange"];
const itemPositions = items.map((item, index) => `${index + 1}: ${item}`);

console.log(itemPositions); // Output: ["1: Apple", "2: Banana", "3: Orange"]
```

## Crucial Best Practices

- **Always use `return`** — The callback function must return a value. If you forget to return something, your new array will be filled with `undefined`.
- **Don't use `map` for simple loops** — If you do not need to create a new array, do not use `map()`. Use `forEach()` or a `for...of` loop instead to save memory.
- **Empty slots are skipped** — If an array has empty holes (sparse array), `map()` will skip them and leave them empty in the new array.
