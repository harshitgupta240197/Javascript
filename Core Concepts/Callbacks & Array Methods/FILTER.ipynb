# `Array.prototype.filter()` in JavaScript

The `Array.prototype.filter()` method in JavaScript creates a new array containing only the elements from the original array that pass a specific condition.

## Key Characteristics

- **Immutability** — It does not change the original array.
- **Shallow Copy** — It returns a brand-new array with the filtered elements.
- **Truthy/Falsy Check** — The callback function must return `true` to keep the item, or `false` to discard it.

## Syntax

```javascript
const newArray = oldArray.filter((element, index, array) => {
  // Return true to keep the element, false to skip it
});
```

- **`element`** — The current item being processed.
- **`index`** *(Optional)* — The index position of the current item.
- **`array`** *(Optional)* — The original source array.

## Common Examples

### 1. Filtering Numbers (Extracting Evens)

```javascript
const numbers = [1, 2, 3, 4, 5, 6];
const evens = numbers.filter(num => num % 2 === 0);

console.log(evens);
// Output: [2, 4, 6]
```

### 2. Filtering Arrays of Objects

You can filter objects based on specific properties, like isolating a targeted product inventory status.

```javascript
const products = [
  { name: 'Laptop', price: 1000, inStock: true },
  { name: 'Phone', price: 500, inStock: false },
  { name: 'Tablet', price: 300, inStock: true }
];

const availableProducts = products.filter(product => product.inStock);

console.log(availableProducts);
// Output: [ { name: 'Laptop', ... }, { name: 'Tablet', ... } ]
```

### 3. Removing Duplicate Values

Using the `index` and reference `array` arguments helps you clean out duplicate values.

```javascript
const duplicates = [1, 2, 2, 3, 4, 4, 5];
const unique = duplicates.filter((value, index, self) => self.indexOf(value) === index);

console.log(unique);
// Output: [1, 2, 3, 4, 5]
```

## Comparison: `filter()` vs `map()` vs `find()`

| Method | What it returns | Modifies original? | Ideal Use Case |
| --- | --- | --- | --- |
| **`filter()`** | A new array with all matching items | No | Extracting a subset of data |
| **`map()`** | A new array of the exact same length | No | Transforming data format |
| **`find()`** | The first single item that matches | No | Looking up a unique item (like an ID) |
