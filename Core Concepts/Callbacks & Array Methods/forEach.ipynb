# Iterating in JavaScript: `.forEach()` vs `for...of`

In JavaScript, you can iterate over arrays and collections using either the `Array.prototype.forEach()` method or the `for...of` loop statement.

## Option 1: The `.forEach()` Method

The `forEach()` method executes a provided callback function once for every element in an array. It is ideal for executing side effects (like logging data, saving to a database, or updating the DOM).

```javascript
const fruits = ['apple', 'banana', 'cherry'];

fruits.forEach((fruit, index) => {
  console.log(`Index ${index}: ${fruit}`);
});

// Output:
// Index 0: apple
// Index 1: banana
// Index 2: cherry
```

### Callback Parameters

The callback function accepts three arguments:

- **`element`** *(Required)* — The current item being processed.
- **`index`** *(Optional)* — The index position of the current item.
- **`array`** *(Optional)* — The original array `forEach()` was called on.

### Critical Limitations

- **No Breaking** — You cannot use `break` or `continue` to stop or skip loops early.
- **Returns Undefined** — It does not return a new array. If you want to transform data into a new array, use `.map()` instead.
- **Async Pitfall** — It executes synchronously and will not wait for `async`/`await` promises.

## Option 2: The `for...of` Loop

Introduced in ES6, `for...of` loops through any iterable object (Arrays, Strings, Maps, Sets, and DOM node lists).

```javascript
const fruits = ['apple', 'banana', 'cherry'];

for (const fruit of fruits) {
  console.log(fruit);
}
```

### Key Advantages

- **Control Statements** — You can use `break`, `continue`, and `return` to control execution flow.
- **Async Friendly** — Works perfectly with `async`/`await` if you need to run asynchronous steps sequentially.

## `.forEach()` vs `for...of`

| Feature | `array.forEach()` | `for...of` |
| --- | --- | --- |
| **Target** | Arrays only | Any iterable (Array, Set, Map, String) |
| **Breaking (`break`/`continue`)** | ❌ No | ✅ Yes |
| **Built-in Index Access** | ✅ Yes (as 2nd parameter) | ❌ No (requires `.entries()`) |
| **Async / Await Support** | ❌ Breaks execution order | ✅ Yes (pauses iterations sequentially) |
