# Variables in JS
#### In JavaScript, variables are named containers used to store data values that can be used and manipulated throughout your code. JavaScript is dynamically typed, meaning you do not need to explicitly declare what type of data (like numbers or strings) a variable will hold; the engine determines it automatically at runtime

# let in JS
#### In JavaScript, the let keyword declares a reassignable, block-scoped variable. Introduced in ES6 (2015), it addresses major scoping and hoisting limitations found in the traditional var keyword
```js
let year = 1985
```

```js
let score = 10;
score = 15; // Works perfectly (Reassignment)

let score = 20; // SyntaxError: Identifier 'score' has already been declared

```
