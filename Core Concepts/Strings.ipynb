# JavaScript String Indexing

In JavaScript, strings use **zero-based indexing**, meaning the first character is at index `0` and the last character is at index `string.length - 1`. The `.length` property returns the total number of code units (characters) in the string.

Here's a quick look at how they work using the word `"Hello"`:

| Character | H | e | l | l | o |
|-----------|---|---|---|---|---|
| Index     | 0 | 1 | 2 | 3 | 4 |

## 📏 The `.length` Property

The `length` property counts every character in the string, including letters, spaces, numbers, and symbols.

```javascript
const text = "Hello World";
console.log(text.length); // Output: 11 (includes the space)
```

## 🔍 Reading Characters by Index

You can read a character at a specific position using three common approaches:

- **Bracket Notation `[]`**: The simplest way to access a character.
- **`.at()` method**: A newer addition that supports negative indices to count backward from the end.
- **`.charAt()` method**: A traditional method that returns an empty string if the index is out of bounds.

```javascript
const str = "JavaScript";

// Bracket Notation
console.log(str[0]);       // Output: "J"

// .at() Method
console.log(str.at(1));    // Output: "a"
console.log(str.at(-1));   // Output: "t" (gets the last character)

// .charAt() Method
console.log(str.charAt(4)); // Output: "S"
```

## 📍 Finding the Index of a Character

If you want to do the opposite and find where a specific character or phrase sits inside a string, use these methods:

- **`.indexOf()`**: Returns the position of the first occurrence. Returns `-1` if not found.
- **`.lastIndexOf()`**: Returns the position of the last occurrence.

```javascript
const phrase = "Banana";
console.log(phrase.indexOf("a"));     // Output: 1
console.log(phrase.lastIndexOf("a")); // Output: 5
console.log(phrase.indexOf("z"));     // Output: -1
```

# String Methods in Javascript
#### In JavaScript, string methods help you inspect, alter, and slice text. Because JavaScript strings are immutable, these methods never change the original string; they always return a brand-new value

# JavaScript String Methods

## 1. Extracting Parts of a String

- `slice(start, end)`: Extracts a section from `start` up to (but not including) `end`. Accepts negative indices to count backward from the end.
- `substring(start, end)`: Similar to `slice()`, but treats negative indices as `0`.

## 2. Finding & Checking Content

- `includes(substring)`: Returns `true` if the string contains the specified text; otherwise `false`.
- `startsWith(substring)` / `endsWith(substring)`: Checks if the string begins or ends with the target text.
- `indexOf(substring)`: Returns the index of the first occurrence of a value, or `-1` if it is not found.
- `lastIndexOf(substring)`: Returns the index of the last occurrence of a value.

## 3. Modifying & Formatting

- `toUpperCase()` / `toLowerCase()`: Converts the entire string to all-caps or all-lowercase.
- `trim()`: Strips whitespace from both the beginning and end of the text.
- `trimStart()` / `trimEnd()`: Removes whitespace from only one specific side.
- `repeat(count)`: Multiplies and returns the string a set number of times.

## 4. Replacing & Splitting

- `replace(search, newText)`: Swaps the first match of a string or regular expression with a new value.
- `replaceAll(search, newText)`: Swaps every single occurrence of a specified value.
- `split(separator)`: Chops a string into an array of substrings based on a delimiter (like a space or comma).

## 5. Inspecting Characters

- `length`: (Property, not a method) Counts the total number of characters.
- `charAt(index)`: Grabs the single character located at that numerical position.
- `at(index)`: Gets the character at an index, allowing negative integers to count back from the final character.


```javascript
const text = "  JavaScript is awesome!  ";

console.log(text.trim().toUpperCase()); // "JAVASCRIPT IS AWESOME!"
console.log(text.includes("Java"));     // true
console.log(text.split(" "));          // ["", "", "JavaScript", "is", "awesome!", "", ""]
```

# JavaScript Template Literals

In JavaScript, template literals are a modern way to create strings using backticks (`` ` ``) instead of standard single or double quotes. They simplify code by supporting embedded expressions, easy multi-line strings, and clean variable formatting without traditional string concatenation.

## Core Features

- **Syntax**: Wrapped in backticks (`` ` ``) instead of single (`'`) or double (`"`) quotes.
- **String Interpolation**: Embed variables and evaluations directly using `${expression}` syntax.
- **Multi-line Strings**: Create new lines directly in the code editor without needing `\n` characters.
- **Easy Quotes**: Use single or double quotes inside the string naturally without escape backslashes.

## Code Examples

### 1. Basic Variable Insertion

Avoid cluttering your code with the `+` operator for concatenation:

```javascript
const user = "Alice";

// Old way
console.log("Hello, " + user + "!");

// With template literals
console.log(`Hello, ${user}!`); // Output: Hello, Alice!
```
