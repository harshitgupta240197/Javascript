# JavaScript String Indexing

In JavaScript, strings use **zero-based indexing**, meaning the first character is at index `0` and the last character is at index `string.length - 1`. The `.length` property returns the total number of code units (characters) in the string.

Here's a quick look at how they work using the word `"Hello"`:

| Character | H | e | l | l | o |
|-----------|---|---|---|---|---|
| Index     | 0 | 1 | 2 | 3 | 4 |

## 📏 The `.length` Property

The `length` property counts every character in the string, including letters, spaces, numbers, and symbols.

```javascript
const text = "Hello World";
console.log(text.length); // Output: 11 (includes the space)
```

## 🔍 Reading Characters by Index

You can read a character at a specific position using three common approaches:

- **Bracket Notation `[]`**: The simplest way to access a character.
- **`.at()` method**: A newer addition that supports negative indices to count backward from the end.
- **`.charAt()` method**: A traditional method that returns an empty string if the index is out of bounds.

```javascript
const str = "JavaScript";

// Bracket Notation
console.log(str[0]);       // Output: "J"

// .at() Method
console.log(str.at(1));    // Output: "a"
console.log(str.at(-1));   // Output: "t" (gets the last character)

// .charAt() Method
console.log(str.charAt(4)); // Output: "S"
```

## 📍 Finding the Index of a Character

If you want to do the opposite and find where a specific character or phrase sits inside a string, use these methods:

- **`.indexOf()`**: Returns the position of the first occurrence. Returns `-1` if not found.
- **`.lastIndexOf()`**: Returns the position of the last occurrence.

```javascript
const phrase = "Banana";
console.log(phrase.indexOf("a"));     // Output: 1
console.log(phrase.lastIndexOf("a")); // Output: 5
console.log(phrase.indexOf("z"));     // Output: -1
```
