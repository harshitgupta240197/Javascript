# Booleans in Javascript
#### In JavaScript, a Boolean is a primitive data type that represents one of two values: true or false. They serve as the foundation for conditional logic, comparison operators, and loop controls
```js
const isPremium = true;
const hasDiscount = false;
```
-----

You can combine or invert boolean conditions using three core logical operators:

| Operator | Name        | Description                                      | Example              |
|----------|-------------|---------------------------------------------------|-----------------------|
| `&&`     | Logical AND | Returns `true` if **both** operands are truthy.   | `true && false → false` |
| `\|\|`   | Logical OR  | Returns `true` if **at least one** operand is truthy. | `true \|\| false → true` |
| `!`      | Logical NOT | **Inverts** the boolean value.                    | `!true → false`       |
