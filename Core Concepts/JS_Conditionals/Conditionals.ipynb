# JavaScript Conditional Statements

Conditional statements in JavaScript control the flow of your program by executing different blocks of code based on whether a condition is true or false. They form the fundamental decision-making engine of any application.

Here is a structured guide to every type of conditional mechanism available in JavaScript:

## 1. The `if` Statement

Executes a single block of code only if the wrapped condition evaluates to `true`.

```javascript
let speed = 70;

if (speed > 60) {
  console.log("Slow down!");
}
```

## 2. The `if...else` Statement

Provides an alternative fallback block of code to run when the initial condition fails.

```javascript
let score = 45;

if (score >= 50) {
  console.log("Passed");
} else {
  console.log("Failed"); // Runs because score is under 50
}
```

## 3. The `if...else if...else` Chain

Evaluates multiple sequential conditions one by one, triggering the first true block it matches.

```javascript
let trafficLight = "yellow";

if (trafficLight === "green") {
  console.log("Go");
} else if (trafficLight === "yellow") {
  console.log("Caution"); // Runs because light is yellow
} else {
  console.log("Stop");
}
```

## 4. The `switch` Statement

Evaluates a single value against multiple defined strict-equality match cases. It is a cleaner alternative to writing long `else if` chains.

- **Crucial Rule**: Always add a `break` statement after each case to prevent the code from automatically executing subsequent cases down the line.

```javascript
let pet = "dog";

switch (pet) {
  case "cat":
    console.log("Meow");
    break;
  case "dog":
    console.log("Woof"); // Runs because pet evaluates to dog
    break;
  default:
    console.log("Unknown animal sound"); // Fallback option
}
```

## 5. Ternary Operator (`? :`)

An ultra-compact shorthand syntax for a simple binary `if...else` statement. It is the only operator in JavaScript that accepts three operands.

- Syntax structure: `condition ? value_if_true : value_if_false`

```javascript
let age = 20;
let access = (age >= 18) ? "Allowed" : "Denied";

console.log(access); // Outputs: "Allowed"
```

## 6. Logical Short-Circuit Operators

You can exploit JavaScript's logical operators to write inline micro-conditionals without using formal syntax blocks.

- **Logical AND (`&&`) Short-Circuit**: If the left side is true, it evaluates and executes the right side.

```javascript
let userLoggedIn = true;
userLoggedIn && console.log("Welcome back!"); // Logs message
```

- **Nullish Coalescing (`??`)**: Provides a specific fallback value if the left expression evaluates explicitly to `null` or `undefined`.

```javascript
let savedUsername = null;
let activeName = savedUsername ?? "Guest";
console.log(activeName); // Outputs: "Guest"
```

## Core Concepts to Keep in Mind

- **Truthy vs Falsy**: JavaScript implicitly converts condition values into booleans. Values like `0`, `""` (empty string), `null`, `undefined`, `NaN`, and `false` are naturally falsy. Everything else is truthy.
- **Strict Equality**: Always favor using triple equals `===` over double equals `==` inside conditions. This ensures you validate both the value and the data type properly, avoiding hidden conversion bugs.

# NESTED IF STATEMENT

```js
let age = 22;
let hasLicense = true;

if (age >= 18) {
    console.log("Age requirement met.");

    // Nested if statement
    if (hasLicense) {
        console.log("Application approved! You can drive the delivery van.");
    } else {
        console.log("Application denied: You need a driving license.");
    }
} else {
    console.log("Application denied: You must be at least 18 years old.");
}

```
----

## 1. Logical AND (`&&`)

The `&&` operator evaluates expressions from left to right and returns the **first falsy value** it encounters. If all values are truthy, it returns the **last** value.

**How it Behaves:**

- Stops executing immediately after finding a falsy value (known as **short-circuit evaluation**).
- Used to execute code only if a prior condition is met.

```javascript
// With booleans
console.log(true && true);       // true
console.log(true && false);      // false

// With other data types (Value Selection)
console.log("Hello" && 0 && 50); // 0 (stops at the first falsy value)
console.log("User" && "Admin");  // "Admin" (all truthy, returns the last one)

// Practical Guard Example
const user = { name: "Alice", premium: true };
// Only runs the right side if the left side is true
user.premium && console.log("Access granted!");
```
-------
## 2. Logical OR (`||`)

The `||` operator evaluates expressions from left to right and returns the **first truthy value** it encounters. If all values are falsy, it returns the **last** value.

**How it Behaves:**

- Short-circuits immediately after finding the first truthy value.
- Commonly used to set **fallback or default values**.

```javascript
// With booleans
console.log(true || false);       // true
console.log(false || false);      // false

// With other data types (Value Selection)
console.log(null || "Guest" || 2); // "Guest" (stops at the first truthy value)
console.log(false || 0 || null);   // null (all falsy, returns the last one)

// Practical Default Example
let userInput = "";
let username = userInput || "Anonymous";
console.log(username);             // "Anonymous" (because "" is falsy)
```
