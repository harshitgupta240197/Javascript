# Truthy and Falsy Values in JavaScript

In JavaScript, truthy and falsy values determine how a non-boolean value is evaluated inside a conditional context (like an `if` statement or loop) through a process called **type coercion**.

JavaScript features a strict, definitive list of exactly **8 falsy values**. Every other value in the language is truthy.

## 🛑 The 8 Falsy Values

When JavaScript expects a boolean and encounters one of these values, it coerces it directly to `false`:

- `false` — The boolean keyword itself.
- `0` — The number zero.
- `-0` — Negative zero.
- `0n` — BigInt zero.
- `""` — An empty string (also `''` or ` `` `).
- `null` — The intentional absence of any object value.
- `undefined` — A variable that has not been assigned a value.
- `NaN` — "Not a Number" (result of a failed math operation).

> **Note:** `document.all` is also treated as falsy, but it is a legacy browser exception rarely seen in modern code.

## The Truthy Values

Everything else evaluates to `true`. This includes several values that often surprise beginners:

- `[]` — Empty arrays are truthy.
- `{}` — Empty objects are truthy.
- `"0"` — Any non-empty string, even if it contains a zero.
- `"false"` — A string containing the word "false".
- `-42` — Any non-zero negative number.
- `function() {}` — Functions, empty or not.

## 📊 Quick Comparison Table

| Value | Type | Context Evaluation |
|-------|------|--------------------|
| `false` | Boolean | 🛑 Falsy |
| `0` or `-0` | Number | 🛑 Falsy |
| `""` | String | 🛑 Falsy |
| `null` | Null | 🛑 Falsy |
| `undefined` | Undefined | 🛑 Falsy |
| `NaN` | Number | 🛑 Falsy |
| `"0"` | String | ✅ Truthy |
| `[]` | Object (Array) | ✅ Truthy |
| `{}` | Object | ✅ Truthy |

## 🛠️ Practical Code Examples

### 1. In Conditional Statements

You don't need to explicitly check if a variable equals `true` or `false`. JavaScript evaluates the truthiness automatically:

```javascript
let username = "Alex"; // Non-empty string (Truthy)
if (username) {
  console.log("Welcome back!"); // This block WILL run
}

let score = 0; // Number zero (Falsy)
if (score) {
  console.log("You have a score!"); // This block WILL NOT run
}
```

### 2. How to Check a Value's Truthiness Explicitly

You can force JavaScript to show you if a value is truthy or falsy using the `Boolean()` function or the double exclamation (`!!`) logical operator:

```javascript
console.log(Boolean([]));      // true
console.log(Boolean(""));      // false
console.log(!!"Hello World");  // true
console.log(!!undefined);      // false
```
