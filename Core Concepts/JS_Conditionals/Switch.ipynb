# JavaScript `switch` Statement

A `switch` statement in JavaScript evaluates an expression and executes a code block associated with a matching `case` value. It provides a cleaner, more readable syntax alternative to lengthy `if...else if...else` chains when checking for exact equality.

## Basic Syntax

```javascript
switch (expression) {
  case value1:
    // Code runs if expression === value1
    break;
  case value2:
    // Code runs if expression === value2
    break;
  default:
    // Code runs if no cases match
}
```

## Key Components

- **`expression`**: The variable or value evaluated exactly once.
- **`case`**: A potential value checked against the expression using strict equality (`===`).
- **`break`**: Exits the switch block immediately. Without it, execution moves into subsequent cases ("fall-through").
- **`default`**: An optional fallback block that executes if no matching case is found.

## Implementation Examples

### 1. Standard Switch Case

This example maps a numerical status code to a user-friendly string message:

```javascript
const statusCode = 404;
let message;

switch (statusCode) {
  case 200:
    message = "Success";
    break;
  case 400:
    message = "Bad Request";
    break;
  case 404:
    message = "Not Found";
    break;
  default:
    message = "Unknown Error";
}

console.log(message); // Output: Not Found
```

### 2. Grouping Cases (Intentional Fall-Through)

You can omit the `break` statement to share a code block among multiple cases:

```javascript
const browser = "Chrome";

switch (browser) {
  case "Chrome":
  case "Firefox":
  case "Edge":
    console.log("This browser is supported.");
    break;
  case "Internet Explorer":
    console.log("This browser is deprecated.");
    break;
  default:
    console.log("Unknown browser.");
}
// Output: This browser is supported.
```

### 3. Matching Conditions (Ranges)

While `switch` traditionally evaluates single values, setting the expression to `true` allows you to evaluate expressions inside individual cases:

```javascript
const score = 85;
let grade;

switch (true) {
  case (score >= 90):
    grade = "A";
    break;
  case (score >= 80):
    grade = "B";
    break;
  default:
    grade = "C";
}

console.log(grade); // Output: B
```
