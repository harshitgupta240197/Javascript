# `==` vs `===` in JavaScript

The main difference between double equals (`==`) and triple equals (`===`) in JavaScript is that `==` compares values after performing type coercion, while `===` compares both the value and the type strictly without conversion.

## Comparison Overview

| Feature | Double Equals (`==`) | Triple Equals (`===`) |
|---|---|---|
| Name | Loose / Abstract Equality | Strict Equality / Identity |
| Type Coercion | Yes (converts types if they differ) | No (returns false if types differ) |
| Performance | Slightly slower (due to conversion) | Slightly faster |
| Best Practice | Avoid it in most scenarios | Highly recommended standard |

## How They Work

### 1. Double Equals (`==`)

If you compare two values of different types, JavaScript implicitly converts one or both values to a common type before executing the comparison.

- `1 == '1'` → `true` (The string `'1'` is converted to the number `1`)
- `true == 1` → `true` (The boolean `true` is converted to the number `1`)
- `null == undefined` → `true` (Special case rule in JavaScript)

### 2. Triple Equals (`===`)

No type conversion happens. If the variable types do not match exactly, the expression immediately evaluates to `false`.

- `1 === '1'` → `false` (Number vs. String)
- `true === 1` → `false` (Boolean vs. Number)
- `null === undefined` → `false` (Different types)
