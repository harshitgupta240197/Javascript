# The JavaScript Math Object

The JavaScript `Math` object is a built-in global object used to perform mathematical operations on numbers.

Unlike many other global objects, `Math` is not a constructor. You cannot create an instance of it using the `new` keyword. All of its properties and methods are static, meaning you call them directly using `Math.methodName()` or `Math.propertyName`.

Detailed documentation on the Math object can be found on the MDN Web Docs Math Reference.


## 📐 Common Math Constants (Properties)

The `Math` object provides several built-in constants for mathematical calculations:

- `Math.PI` — Pi ratio (≈ 3.14159)
- `Math.E` — Euler's number (≈ 2.718)
- `Math.SQRT2` — Square root of 2 (≈ 1.414)
- `Math.LN10` — Natural logarithm of 10 (≈ 2.302)

## 🔢 Essential Math Methods

The methods can be broadly categorized into rounding, powers, roots, and basic utility functions.

### 1. Rounding Numbers

- `Math.round(x)` — Rounds to the nearest integer.
- `Math.ceil(x)` — Rounds up to the next integer.
- `Math.floor(x)` — Rounds down to the previous integer.
- `Math.trunc(x)` — Removes the decimal part entirely.

```javascript
Math.round(4.5); // 5
Math.ceil(4.1);  // 5
Math.floor(4.9); // 4
Math.trunc(4.9); // 4
```

### 2. Powers and Roots

- `Math.pow(x, y)` — Returns x to the power of y.
- `Math.sqrt(x)` — Returns the square root of x.
- `Math.cbrt(x)` — Returns the cube root of x.

```javascript
Math.pow(2, 3); // 8
Math.sqrt(16);  // 4
```

### 3. Absolute, Minimum, and Maximum Values

- `Math.abs(x)` — Returns the absolute (positive) value.
- `Math.min(...values)` — Returns the lowest value among arguments.
- `Math.max(...values)` — Returns the highest value among arguments.

```javascript
Math.abs(-7);        // 7
Math.max(1, 9, 3, 5); // 9
Math.min(1, 9, 3, 5); // 1
```

### 4. Generating Random Numbers

The `Math.random()` method returns a pseudo-random decimal number between 0 (inclusive) and 1 (exclusive). To get a random integer within a specific range, you combine it with `Math.floor()`:

```javascript
// Generate a random integer between 1 and 10 (inclusive)
let randomNum = Math.floor(Math.random() * 10) + 1;
```
