# Javascript
#### JavaScript (JS) is a lightweight, interpreted programming language used primarily to make web pages interactive and dynamic. While HTML structures a page and CSS styles it, JavaScript controls how the website behaves, handling elements like pop-ups, calculations, animations, and data updates. Today, it runs both inside the web browser (client-side) and on servers via Node.js (server-side)
