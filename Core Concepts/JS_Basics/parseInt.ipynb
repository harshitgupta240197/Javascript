# The `parseInt()` Function in JavaScript

The JavaScript `parseInt()` function parses a string argument and returns an integer of the specified radix (base). If the first character cannot be converted to a number, it returns `NaN` (Not a Number). [[1]](https://developer.mozilla.org/en-US/docs/Web/JavaScript/Reference/Global_Objects/parseInt) [[2]](https://www.geeksforgeeks.org/javascript/javascript-number-parseint-method/) [[3]](https://developer.mozilla.org/en-US/docs/Web/JavaScript/Reference/Global_Objects/parseInt)

## Syntax

```javascript
parseInt(string, radix);
```

- **`string`**: The value you want to parse. If it is not a string, JavaScript converts it to one. Leading whitespace is ignored.
- **`radix`**: An optional integer between `2` and `36` representing the mathematical numeral system (e.g., `10` for decimal, `2` for binary, `16` for hexadecimal). [[1]](https://medium.com/@cristi.nord/javascript-parseint-c6b2a271f153) [[2]](https://masteringjs.io/tutorials/fundamentals/parseint) [[3]](https://www.w3schools.com/jsref/jsref_parseint.asp) [[4]](https://developer.mozilla.org/en-US/docs/Web/JavaScript/Reference/Global_Objects/parseInt)

## Quick Code Examples

```javascript
// Basic usage (Defaults to base 10 in most modern cases)
parseInt("42");        // Returns 42

// Stops parsing at the first non-numeric character
parseInt("42px");      // Returns 42
parseInt("42.99");     // Returns 42 (does not round, just drops decimals)

// Returns NaN if the first character isn't a number
parseInt("abc42");     // Returns NaN

// Specifying a radix (Binary)
parseInt("1010", 2);   // Returns 10
```

## Key Behaviors & Quirks

- **Aggressive Parsing**: It reads from left to right and stops evaluating as soon as it hits a character it does not recognize in the specified radix. [[1]](https://www.youtube.com/watch?v=TjaNRrNfHjY&vl=en) [[2]](https://www.geeksforgeeks.org/javascript/javascript-number-parseint-method/)
- **Always Provide the Radix**: If you omit the radix, JavaScript guesses based on the string format. For example, strings starting with `0x` are parsed as hexadecimal (base 16). Always explicitly provide `10` for standard decimal numbers to prevent unexpected bugs. [[1]](https://developer.mozilla.org/en-US/docs/Web/JavaScript/Reference/Global_Objects/parseInt) [[2]](https://stackoverflow.com/questions/1133770/how-can-i-convert-a-string-to-an-integer-in-javascript) [[3]](https://masteringjs.io/tutorials/fundamentals/parseint)
- **`parseInt()` vs `Number()`**: While `parseInt("42px")` extracts `42`, the `Number("42px")` function will strictly return `NaN` because the entire string is not a pure number.
