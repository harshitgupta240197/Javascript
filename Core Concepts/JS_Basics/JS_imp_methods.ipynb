# `console.log()`, `alert()`, and `prompt()` in JavaScript

In JavaScript, `console.log()`, `alert()`, and `prompt()` are three foundational methods used for displaying information, debugging, and collecting user input.

## Quick Comparison

| Method | Purpose | User Interaction | Returns | Blocks Execution? |
|---|---|---|---|---|
| `console.log()` | Development / Debugging | None (Invisible to user) | `undefined` | No |
| `alert()` | Showing notifications | Clicks "OK" | `undefined` | Yes (Pauses script) |
| `prompt()` | Getting text input | Types text + OK/Cancel | String or `null` | Yes (Pauses script) |

## 1. `console.log()` — The Developer's Tool

Prints data directly to the browser's developer console. It is strictly used for logging, testing, and debugging behind the scenes.

```javascript
let framework = "JavaScript";
console.log("Debugging information:", framework);
```

## 2. `alert()` — The Notification Pop-up

Displays a modal dialog box containing a message and a single "OK" button. It forces the user to acknowledge the message before they can interact with the webpage.

```javascript
alert("Form submitted successfully!");
```

## 3. `prompt()` — The Input Pop-up

Opens a modal dialog box with a message, an input text field, and "OK" / "Cancel" buttons.

- Clicking OK returns the entered text as a string.
- Clicking Cancel or pressing Esc returns `null`.

```javascript
// Syntax: prompt(message, default_text)
let age = prompt("Please enter your age:", "21");

if (age !== null) {
    console.log("The user is " + age + " years old.");
} else {
    console.log("User cancelled the input.");
}
```
