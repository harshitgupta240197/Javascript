# JavaScript `while` Loops

A `while` loop in JavaScript executes a block of code repeatedly as long as a specified condition evaluates to `true`. It is primarily used when the total number of iterations is not known in advance.

## Basic Syntax

The loop checks the condition **before** executing the code block. If the condition starts as `false`, the code block will not run at all.

```javascript
while (condition) {
  // code block to be executed
}
```

## Code Example

Here is a basic loop that counts from 0 to 2:

```javascript
let i = 0; // 1. Initialize the counter
while (i < 3) { // 2. Check the condition
  console.log(i);
  i++; // 3. Update the counter to prevent an infinite loop
}
```

## Critical Components

Every standard `while` loop requires three parts to function safely:

- **Initializer**: A variable set before the loop starts to track progress.
- **Condition**: A boolean expression tested before every single iteration.
- **Update Expression**: Code inside the block that alters the initializer so the condition eventually becomes false. Missing this creates an infinite loop, which can freeze your browser or application.

## While vs. Do...While

JavaScript also offers a variation called the `do...while` loop.

| Loop Type    | When Condition is Checked | Minimum Iterations               |
|--------------|---------------------------|-----------------------------------|
| `while`      | Before the block runs     | 0                                  |
| `do...while` | After the block runs      | 1 (Always runs at least once)      |

```javascript
// This will NOT run because the condition is immediately false
let x = 10;
while (x < 5) {
  console.log("While:", x);
  x++;
}

// This WILL run exactly once despite the condition being false
let y = 10;
do {
  console.log("Do...While:", y);
  y++;
} while (y < 5);
```

## Loop Control Statements

You can alter the behavior of a loop from the inside using these keywords:

- **`break`**: Immediately terminates the loop and jumps to the next line of code outside it.
- **`continue`**: Skips the rest of the current iteration and jumps directly to checking the condition again.

## Guessing number game
```js
// Write a number guessing game

const maximum = parseInt(prompt('Enter the maximum number: '))

while (!maximum) {
    maximum = parseInt(prompt('Enter the maximum number: '))  
}

const targetNum = Math.floor(Math.random() * maximum) + 1;

let guess = prompt('Enter your first guess!')
let attempts = 1;

while (parseInt(guess) !== targetNum) {
    if (guess === 'q') break;
    attempts++
    if (guess > targetNum) {
        guess = prompt('Too High! Enter a new guess')
    } else {
        guess = prompt('Too Low! Enter a new guess');
    }
}

if (guess === 'q') {
    console.log('OK, you quit');
} else {
    console.log('Congrats You Win!');
    console.log(`You got it! It took you ${attempts} guesses`);
}
```
