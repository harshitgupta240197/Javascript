# The `break` Keyword in JavaScript

The `break` keyword in JavaScript immediately terminates the execution of a loop or a `switch` statement. When the program encounters `break`, it instantly exits the loop and moves directly to the next line of code written below that loop. [[1]](https://www.w3schools.com/js/js_break.asp) [[2]](https://www.w3schools.com/jsref/jsref_break.asp)

## Code Examples

### 1. In a `for` Loop

You can stop a standard loop early once a specific condition is met. [[1]](https://www.youtube.com/watch?v=QSuTH0C_3_Y&t=50)

```javascript
for (let i = 1; i <= 10; i++) {
  if (i === 4) {
    break; // Exits the loop entirely when i equals 4
  }
  console.log(i);
}
// Output: 1, 2, 3
```

### 2. In a `while` Loop

Useful for stopping loops that might otherwise run infinitely. [[1]](https://www.udacity.com/blog/javascript-break-and-continue/)

```javascript
let count = 0;

while (true) {
  console.log(count);
  count++;

  if (count === 3) {
    break; // Halts the infinite loop
  }
}
// Output: 0, 1, 2
```

## Important Distinctions & Rules

- **`break` vs. `continue`**: While `break` stops the entire loop, the `continue` keyword only skips the current iteration and jumps directly to the next one. [[1]](https://www.youtube.com/watch?v=SyzffZJzLNo) [[2]](https://flaviocopes.com/courses/javascript/break-loop/)
- **Nested Loops**: A standard `break` statement only exits the innermost loop that contains it. If you need to break out of an outer loop from inside an inner loop, you must use a [labeled statement](https://developer.mozilla.org/en-US/docs/Web/JavaScript/Reference/Statements/break). [[1]](https://developer.mozilla.org/en-US/docs/Web/JavaScript/Reference/Statements/break) [[2]](https://developer.mozilla.org/en-US/docs/Web/JavaScript/Guide/Loops_and_iteration)
- **Array Methods**: You cannot use the `break` keyword inside modern array iteration methods like `.forEach()`, `.map()`, or `.filter()`. If you need early termination capability with arrays, use a standard `for...of` loop instead.
