# JavaScript `array.reduce()`

The `array.reduce()` method executes a user-supplied "reducer" callback on each element of an array to combine them into a single value. It iterates left to right, passing the return value from the previous element's calculation into the next iteration.

## Syntax

```javascript
array.reduce((accumulator, currentValue, currentIndex, array) => {
  // Logic here
}, initialValue);
```

## Core Parameters

| Parameter | Required | Description |
| --- | --- | --- |
| `accumulator` | Yes | The running total or accumulated value returned from the previous iteration. |
| `currentValue` | Yes | The current element being processed. |
| `currentIndex` | Optional | The index of the current element. |
| `array` | Optional | The original array `reduce()` was called upon. |
| `initialValue` | Optional | Value used as the accumulator on the first call. If omitted, the first element becomes the initial accumulator and iteration starts at the second element. |

## Common Use Cases

### 1. Summing an array of numbers

The most basic application — accumulating numbers.

```javascript
const numbers = [10, 20, 30, 40];

const totalSum = numbers.reduce((accumulator, currentValue) => {
  return accumulator + currentValue;
}, 0);

console.log(totalSum); // Output: 100
```

### 2. Flattening a multi-dimensional array

Merge a nested array into a single-dimensional array.

```javascript
const nestedArray = [[1, 2], [3, 4], [5, 6]];

const flattened = nestedArray.reduce((accumulator, currentValue) => {
  return accumulator.concat(currentValue);
}, []);

console.log(flattened); // Output: [1, 2, 3, 4, 5, 6]
```

### 3. Counting occurrences (frequency map)

Set `initialValue` to an empty object `{}` to group and tally values.

```javascript
const fruits = ['apple', 'banana', 'apple', 'orange', 'banana', 'apple'];

const fruitCounts = fruits.reduce((accumulator, fruit) => {
  accumulator[fruit] = (accumulator[fruit] || 0) + 1;
  return accumulator;
}, {});

console.log(fruitCounts);
// Output: { apple: 3, banana: 2, orange: 1 }
```

### 4. Grouping objects by property

Organize data sets based on specific criteria.

```javascript
const team = [
  { name: 'Alice', role: 'Developer' },
  { name: 'Bob', role: 'Designer' },
  { name: 'Charlie', role: 'Developer' }
];

const groupedByRole = team.reduce((accumulator, person) => {
  const role = person.role;
  if (!accumulator[role]) {
    accumulator[role] = [];
  }
  accumulator[role].push(person.name);
  return accumulator;
}, {});

console.log(groupedByRole);
// Output: { Developer: ['Alice', 'Charlie'], Designer: ['Bob'] }
```

## Best Practices

- **Always return the accumulator.** If you forget to `return` it inside the callback, the next iteration receives `undefined`.
- **Provide an initial value.** Passing `0`, `[]`, or `{}` prevents errors on empty arrays and keeps the accumulator's type consistent.
- **Weigh readability against cleverness.** `reduce()` is versatile, but a `map()`/`filter()` combination or a plain `for...of` loop is often easier for a team to read and maintain.
