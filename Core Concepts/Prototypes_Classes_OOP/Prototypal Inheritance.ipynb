# Prototypal Inheritance in JavaScript

## Definition

**Prototypal inheritance** is a core JavaScript feature where objects directly inherit properties and methods from other objects. Unlike traditional class-based languages (like Java or C++) that use classes as blueprints to create instances, JavaScript links instances together via an internal property known as `[[Prototype]]`.

## The Prototype Chain

Every JavaScript object contains a hidden reference to another object called its prototype. When you attempt to access a property or execute a method on an object, the JavaScript engine executes a specific lookup process:

1. **Local Check** — It looks for the property directly on the object itself.
2. **Chain Traversal** — If not found locally, the engine searches the object's `[[Prototype]]`.
3. **Termination** — This search moves recursively up the chain until the property is found, or it reaches `null` (the ultimate end of the chain, which has no prototype).

> **Note:** `__proto__` is the (informal, now-standardized) accessor that lets you *see* an object's `[[Prototype]]` in the console or code. `prototype` is a different thing — it's a property that exists only on **functions**, and it's the object that becomes the `[[Prototype]]` of anything created with `new FunctionName()`. Mixing these two up is a very common source of confusion.

## Practical Implementation Methods

There are three primary ways to work with prototypal inheritance in JavaScript.

### 1. Modern Approach: `Object.create()`

This is the cleanest method to link a new object directly to an existing parent object.

```javascript
const animal = {
  makeSound() {
    console.log(`${this.name} makes a sound.`);
  }
};

// Create a new object with 'animal' as its prototype
const dog = Object.create(animal);
dog.name = "Buddy";

dog.makeSound(); // Output: "Buddy makes a sound."
```

### 2. Constructor Functions (Legacy / Behind-the-Scenes)

Before ES6, constructor functions combined with the `new` keyword were used to scale object creation.

```javascript
function Animal(name) {
  this.name = name;
}

// Attach a method to the constructor's prototype property
Animal.prototype.makeSound = function() {
  console.log(`${this.name} makes a sound.`);
};

const dog = new Animal("Buddy");
dog.makeSound(); // Inherited from Animal.prototype
```

### 3. ES6 Classes (Syntactic Sugar)

The `class` keyword provides a cleaner syntax that mimics traditional object-oriented languages. However, underneath the hood, it still compiles to prototypal delegation.

```javascript
class Animal {
  constructor(name) {
    this.name = name;
  }
  makeSound() {
    console.log(`${this.name} makes a sound.`);
  }
}

class Dog extends Animal {}
const dog = new Dog("Buddy");
dog.makeSound();
```

## Key Characteristics to Remember

- **Differential Inheritance** — Methods and properties are not copied from parent to child. They remain in memory on the prototype object once, conserving memory.
- **The Behavior of `this`** — When an inherited function is executed, `this` points directly to the calling object, not the prototype object where the function is defined.
- **Read vs. Write Execution** — Property lookups navigate up the chain, but writing or deleting data properties (`dog.name = "Max"`) always alters the target object directly, never its prototype.
- **Runtime Mutation** — Prototypes can be swapped or modified dynamically at runtime using `Object.setPrototypeOf()`, altering downstream behavior immediately.

## Useful Prototype-Related Tools

A few methods that come up constantly once you start working with prototypes in practice — not in the original notes, but worth having alongside them:

```javascript
// Check if a property exists directly on the object (not the chain)
dog.hasOwnProperty("name");       // true
dog.hasOwnProperty("makeSound");  // false — it's inherited

// Get an object's prototype directly
Object.getPrototypeOf(dog);       // -> animal

// Check if an object exists anywhere in another's prototype chain
animal.isPrototypeOf(dog);        // true
```

`hasOwnProperty` is especially useful in `for...in` loops, since `for...in` iterates over inherited enumerable properties too — filtering with `hasOwnProperty` keeps you to just the object's own keys.

## Prototypal vs. Classical Inheritance

| | Classical (Java, C++) | Prototypal (JavaScript) |
|---|---|---|
| Blueprint | Classes | Objects |
| New instance created from | A class definition | An existing object |
| Inheritance mechanism | Copy behavior into subclass | Link/delegate to another object |
| Can change after creation? | No — structure fixed at compile time | Yes — prototype chain is mutable at runtime |
| `class` keyword in JS | N/A | Syntactic sugar over the above |

## Quick Recap

| Concept | One-line summary |
|---|---|
| `[[Prototype]]` | Internal link every object has to another object |
| Prototype chain | The lookup path followed when a property isn't found locally |
| `Object.create()` | Cleanest modern way to set an object's prototype explicitly |
| Constructor + `new` | Pre-ES6 way to create linked instances at scale |
| `class` / `extends` | Syntactic sugar; still prototypal delegation underneath |
| `hasOwnProperty()` | Checks if a property is on the object itself, not inherited |
