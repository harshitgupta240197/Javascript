# Axios

Axios is a popular, promise-based HTTP client library for JavaScript. It allows you to send asynchronous network requests to REST endpoints and works identically in both web browsers and Node.js environments.

## Installation

Install it into your project using a package manager like NPM:

```bash
npm install axios
```

Alternatively, include it in an HTML file using a CDN link:

```html
<script src="https://jsdelivr.net"></script>
```

## Core HTTP Request Examples

Axios natively integrates with `async/await` syntax for clean asynchronous code.

### 1. GET Request (Fetching Data)

```javascript
import axios from 'axios';

async function getUser() {
  try {
    // Axios automatically parses JSON data into a JavaScript object
    const response = await axios.get('https://example.com');
    console.log(response.data);
  } catch (error) {
    console.error('Request failed:', error.message);
  }
}
```

### 2. POST Request (Sending Data)

```javascript
async function createUser() {
  try {
    const payload = { name: 'Alice', role: 'Developer' };
    const response = await axios.post('https://example.com', payload);
    console.log('Created:', response.data);
  } catch (error) {
    console.error('Failed to create user:', error.message);
  }
}
```

## Axios vs. Native Fetch API

While modern browsers include the native `fetch()` API, many developers prefer Axios because it handles a lot of boilerplate code automatically.

| Feature | Axios | Fetch API |
| --- | --- | --- |
| JSON Data | Automatically parsed into an object | Requires an explicit `.json()` step |
| Error Handling | Throws errors for 4xx and 5xx statuses | Only throws errors on network failure |
| Timeouts | Supported natively via a `timeout` property | Requires a separate `AbortController` setup |
| Interceptors | Built-in request and response modifiers | Not natively supported |
| Backwards Compatibility | Wide support, including legacy browsers | Only available in modern browsers |

## Advanced Features

### Global Configuration and Instances

You can configure a custom reusable instance of Axios to specify default headers and base URLs:

```javascript
const apiClient = axios.create({
  baseURL: 'https://example.com',
  timeout: 5000, // Aborts if the request takes longer than 5 seconds
  headers: { 'Authorization': 'Bearer YOUR_TOKEN' }
});

// Uses the base URL automatically: https://example.com/products
apiClient.get('/products');
```

### Interceptors

Interceptors let you execute code or modify requests right before they leave, or modify responses before they are processed by your code.

```javascript
// Automatically attach an auth token to every outgoing request
axios.interceptors.request.use((config) => {
  config.headers.Authorization = `Bearer ${localStorage.getItem('token')}`;
  return config;
});
```
