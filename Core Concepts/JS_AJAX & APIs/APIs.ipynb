# APIs in JavaScript

An **API** (Application Programming Interface) is a set of features and tools that allows your code to communicate with other software, browsers, or external servers.

JavaScript APIs generally fall into two categories: **Web/Browser APIs** (built directly into the browser) and **Third-Party APIs** (hosted by external services).

---

## 1. Browser APIs (Built-in)

Native features provided by the browser environment to help you interact with the system or the web page.

- **DOM API** — Modifies HTML structure, text elements, and CSS styles.
- **Fetch API** — Makes network requests to send or retrieve data from servers asynchronously.
- **Storage APIs** — Saves data on the user's browser using `localStorage` or `sessionStorage`.
- **Geolocation API** — Retrieves the physical geographic coordinates of the user's device.

---

## 2. Third-Party APIs (External)

These require an HTTP request over the internet to fetch data from an external platform's database.

- **Weather data** — OpenWeatherMap API
- **Mapping services** — Google Maps API
- **Payment gateways** — Stripe API

---

## How to Call an API Using Fetch

The modern standard for interacting with external APIs is the promise-based `fetch()` function paired with `async` / `await` syntax.

```javascript
// Step 1: Define an async function
async function getUserData() {
  try {
    // Step 2: Make the network request (GET)
    const response = await fetch('https://jsonplaceholder.typicode.com/users/1');

    // Step 3: Verify the response is successful
    if (!response.ok) {
      throw new Error(`HTTP error! Status: ${response.status}`);
    }

    // Step 4: Parse the response payload into a JSON object
    const data = await response.json();
    console.log(data.name);

  } catch (error) {
    // Step 5: Handle network failures or errors
    console.error('Failed to fetch data:', error);
  }
}

// Execute the call
getUserData();
```

---

## Common HTTP Methods Used with APIs

| Method | Purpose |
| --- | --- |
| `GET` | Retrieves existing information from a database (default method for `fetch`). |
| `POST` | Submits new data to a server (e.g. creating a new user profile). |
| `PUT` / `PATCH` | Updates existing database records — `PUT` replaces the whole record, `PATCH` modifies part of it. |
| `DELETE` | Removes data from a remote server. |

Anything other than `GET` requires an options object as the second argument to `fetch`:

```javascript
await fetch('https://example.com/api/users', {
  method: 'POST',
  headers: { 'Content-Type': 'application/json' },
  body: JSON.stringify({ name: 'Harshit' })
});
```
