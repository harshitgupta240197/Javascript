# HTTP Methods in JavaScript

HTTP methods (also known as **HTTP verbs**) specify the intended action when making network requests to a server. In the browser they are primarily executed using the **Fetch API**.

---

## Core HTTP Methods

| Method | Primary purpose | Safe? | Idempotent? |
| --- | --- | --- | --- |
| `GET` | Retrieves data from a server without modifying it. | Yes | Yes |
| `POST` | Submits new data to create a resource or trigger side effects. | No | No |
| `PUT` | Replaces an entire resource with the new payload. | No | Yes |
| `PATCH` | Partially modifies an existing resource. | No | No |
| `DELETE` | Removes a resource from the server. | No | Yes |

- **Safe** — the request does not alter server state.
- **Idempotent** — making the exact same request multiple times produces the same end state.

> **Why idempotency matters:** if a `PUT` request times out, you can safely retry it. Retrying a `POST` may create a duplicate record. This is the whole reason payment APIs use idempotency keys.

## Implementation Examples using Fetch API

### 1. GET Request (default)

Retrieves information. Parameters are appended directly to the URL string.

```javascript
async function getData() {
  try {
    const response = await fetch('https://example.com/api/users');
    if (!response.ok) throw new Error('Network error');
    const data = await response.json();
    console.log(data);
  } catch (error) {
    console.error('Failed:', error);
  }
}
```

---

### 2. POST Request

Sends data inside the body payload to create something new. Always stringify objects using `JSON.stringify()`.

```javascript
async function createUser() {
  const response = await fetch('https://example.com/api/users', {
    method: 'POST',
    headers: { 'Content-Type': 'application/json' },
    body: JSON.stringify({ name: 'Alex', role: 'Admin' })
  });
  return response.json();
}
```

---

### 3. PUT Request

Overwrites the targeted resource entirely.

```javascript
async function updateAllUserData(userId) {
  const response = await fetch(`https://example.com/api/users/${userId}`, {
    method: 'PUT',
    headers: { 'Content-Type': 'application/json' },
    body: JSON.stringify({ name: 'Alex', role: 'Editor' }) // Replaces entire entry
  });
  return response.json();
}
```

---

### 4. PATCH Request

Updates only the specified properties, leaving others untouched.

```javascript
async function updateRole(userId) {
  const response = await fetch(`https://example.com/api/users/${userId}`, {
    method: 'PATCH',
    headers: { 'Content-Type': 'application/json' },
    body: JSON.stringify({ role: 'Superadmin' }) // Only alters the role
  });
  return response.json();
}
```

---

### 5. DELETE Request

Instructs the server to remove the specific item. Typically requires no body content.

```javascript
async function deleteUser(userId) {
  const response = await fetch(`https://example.com/api/users/${userId}`, {
    method: 'DELETE'
  });
  return response.ok;
}
```

> Note that `DELETE` often returns `204 No Content`. Calling `.json()` on an empty body throws — check `response.status` or `response.ok` instead.

## Alternative Request Environments

- **Axios** — a popular third-party package used in both browser and Node.js code. It cuts down boilerplate by auto-serialising and auto-parsing JSON, and it rejects on 4xx/5xx status codes rather than resolving the way `fetch` does.
- **Node.js native** — server-side applications issue or handle requests through the built-in `http` module, global `fetch` (available natively since Node 18), or frameworks like Express.
