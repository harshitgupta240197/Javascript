# JSON in JavaScript

**JSON** (JavaScript Object Notation) is a lightweight text format used to store and exchange structured data. In JavaScript, the global `JSON` object provides built-in methods to convert data between text strings and native objects.

---

## The Two Core Methods

| Method | Direction | Purpose |
| --- | --- | --- |
| `JSON.parse()` | String → Object | Converts a valid JSON string into a usable JavaScript object. |
| `JSON.stringify()` | Object → String | Converts a JavaScript object or value into a JSON-formatted string. |

---

## 1. Converting a JSON String to a JavaScript Object

When you fetch data from a web server or an API, it arrives as a text string. Use `JSON.parse()` to turn it into a JavaScript object so you can access its properties.

```javascript
// A JSON string received from a server
const jsonString = '{"name": "Alice", "age": 25, "isAdmin": true}';

// Convert string to a JavaScript object
const userObj = JSON.parse(jsonString);

// Access properties normally
console.log(userObj.name); // Output: Alice
console.log(userObj.age);  // Output: 25
```

> **Note:** If the string is invalid JSON, `JSON.parse()` throws a `SyntaxError`. Wrap it in a `try...catch` block to prevent your script from crashing.

---

## 2. Converting a JavaScript Object to a JSON String

When sending data to a server or saving it to local storage, you must convert your JavaScript object into a text string using `JSON.stringify()`.

```javascript
// A standard JavaScript object
const profile = {
  username: "DevJS",
  skills: ["JavaScript", "HTML"],
  active: false
};

// Convert object to a JSON string
const jsonText = JSON.stringify(profile);

console.log(jsonText);
// Output: '{"username":"DevJS","skills":["JavaScript","HTML"],"active":false}'

console.log(typeof jsonText); // Output: "string"
```

`stringify` also takes an optional third argument for indentation, which makes output readable when debugging:

```javascript
JSON.stringify(profile, null, 2);
// {
//   "username": "DevJS",
//   ...
// }
```

---

## Strict Differences: JavaScript Objects vs. JSON Syntax

While JSON is derived from JavaScript object syntax, it enforces much stricter formatting rules:

- **Double quotes required** — Property keys and string values must use double quotes (`"key"`). Single quotes are invalid.
- **No trailing commas** — Strictly forbidden after the last property or array item.
- **Data types only** — JSON supports strings, numbers, booleans, objects, arrays, and `null`.
- **Unsupported types** — Functions, `undefined`, `NaN`, `Infinity`, and comments are prohibited. If you pass an object containing a function to `JSON.stringify()`, that property is silently dropped.

### The `Date` exception

`Date` objects aren't a JSON type, but they don't error either — `stringify` converts them to an ISO string:

```javascript
JSON.stringify({ created: new Date() });
// '{"created":"2026-08-21T09:30:00.000Z"}'
```

The catch is that this is a one-way trip. `JSON.parse()` gives you back a plain string, not a `Date`, so you have to rehydrate it yourself with `new Date(obj.created)`. This is a common source of bugs when round-tripping data through `localStorage`.
