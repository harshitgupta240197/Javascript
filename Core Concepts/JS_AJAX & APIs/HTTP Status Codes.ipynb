# HTTP Status Codes

HTTP status codes are standard three-digit responses returned by a web server to indicate the outcome of a client's request. Standardised by the **IETF**, they are split into five families based on the first digit.

| Family | Meaning |
| --- | --- |
| `1xx` | Informational — request received, continue |
| `2xx` | Success — received, understood, processed |
| `3xx` | Redirection — further action needed |
| `4xx` | Client error — the request was wrong |
| `5xx` | Server error — the request was fine, the server broke |

---

## 1xx — Informational

The server received the request and is processing it; the client should continue.

| Code | Meaning |
| --- | --- |
| `100 Continue` | The initial request part is accepted; proceed with the rest. |
| `101 Switching Protocols` | The server is changing protocols (e.g. upgrading to WebSockets). |

---

## 2xx — Success

The request was successfully received, understood, and processed.

| Code | Meaning |
| --- | --- |
| `200 OK` | The standard successful response for fetching or changing data. |
| `201 Created` | The request succeeded and a new resource was generated. |
| `202 Accepted` | The request is queued for asynchronous processing but not yet finished. |
| `204 No Content` | Action successful, but the response deliberately contains no body. |

> **Practical note:** a well-behaved API returns `201` from a `POST` that creates something, and `204` from a `DELETE`. Calling `.json()` on a `204` response throws — check `response.ok` instead.

---

## 3xx — Redirection

Further navigation must be taken by the client to fulfil the request.

| Code | Meaning |
| --- | --- |
| `301 Moved Permanently` | The resource has permanently been assigned a new URL. |
| `302 Found` | The resource temporarily resides at a different URL. |
| `304 Not Modified` | The cached version matches the server; no download required. |

---

## 4xx — Client Error

The request contains incorrect syntax, missing data, or bad authorisation from the client side.

| Code | Meaning |
| --- | --- |
| `400 Bad Request` | Malformed syntax, invalid parameters, or validation failures. |
| `401 Unauthorized` | Missing or invalid authentication credentials. |
| `403 Forbidden` | Authenticated, but lacking permission for this resource. |
| `404 Not Found` | The server cannot locate the resource at that URL. |
| `429 Too Many Requests` | Rate limit triggered by sending too many requests too quickly. |

> **401 vs 403** — the names are misleading. `401` means *"I don't know who you are"* (authentication). `403` means *"I know exactly who you are, and no"* (authorisation). Worth pinning down, since it's a standard interview question and a common source of confused API design.

---

## 5xx — Server Error

The server failed to fulfil an otherwise valid request due to an internal technical issue.

| Code | Meaning |
| --- | --- |
| `500 Internal Server Error` | Generic catch-all for an unexpected crash or bug. |
| `502 Bad Gateway` | A proxy received an invalid response from the upstream application. |
| `503 Service Unavailable` | The server is temporarily overloaded or down for maintenance. |
| `504 Gateway Timeout` | The proxy timed out waiting for the application to respond. |

> **Debugging heuristic:** `4xx` means fix your request. `5xx` means the request was fine and the problem is on the server side — stop rewriting your fetch call and go read the server logs.

---

## Handling Status Codes in `fetch`

`fetch` does **not** reject on `4xx` or `5xx` — those still resolve successfully. Only network failure rejects. This is why the explicit check is mandatory:

```javascript
const response = await fetch(url);

if (!response.ok) {                     // true for any 2xx, false otherwise
  throw new Error(`HTTP ${response.status}: ${response.statusText}`);
}
```
