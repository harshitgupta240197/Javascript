# JavaScript Functions

In JavaScript, a function is a reusable block of code designed to perform a specific task. It eliminates code repetition by allowing you to write logic once and execute it multiple times throughout your application.

## Ways to Define Functions

JavaScript offers multiple syntaxes to define functions depending on your architectural needs:

| Type                 | Syntax Example                    | Key Characteristic                                          |
|----------------------|-----------------------------------|-------------------------------------------------------------|
| Function Declaration | `function greet() { ... }`        | Hoisted; can be called before it is defined in the code.    |
| Function Expression  | `const greet = function() { ... }`| Not hoisted; treated as a variable value.                   |
| Arrow Function       | `const greet = () => { ... }`     | Shorter syntax; does not have its own `this` context.       |
| IIFE                 | `(function() { ... })()`          | Runs automatically as soon as it is defined.                |

## Core Anatomy of a Function

A standard function consists of **parameters** (inputs), a **body** (logic), and a **return statement** (output).

```javascript
// 1. Defining the function with parameters (placeholders)
function calculateTotal(price, tax) {
  const total = price + (price * tax);
  return total; // 2. Specifying the output value
}

// 3. Invoking (calling) the function with real arguments
const finalAmount = calculateTotal(100, 0.15);
console.log(finalAmount); // Outputs: 115
```

- **Parameters vs. Arguments**: Parameters are the variables listed in the function definition (`price`, `tax`). Arguments are the actual values passed to the function when calling it (`100`, `0.15`).
- **The Return Statement**: The `return` keyword stops the function's execution and sends a value back to the caller. If no `return` statement is specified, the function returns `undefined` by default.

## Advanced Concepts

### First-Class Citizens

Functions in JavaScript are **First-Class Objects**. This means they can be treated like any other variable. You can:

- Pass them as arguments to other functions (known as **callbacks**).
- Return them from other functions.
- Assign them to variables or object properties (known as **methods**).

### Scope and Closures

- **Local Scope**: Variables declared inside a function cannot be accessed from outside the function.
- **Closures**: A function remembers and can access variables from its outer (parent) lexical scope, even after the parent function has finished executing.

## Specialized Behaviors

Beyond regular functions, JavaScript includes specific engines for complex operations:

- **Async Functions**: Uses the `async`/`await` syntax to handle asynchronous operations cleanly without blocking execution.
- **Generator Functions**: Uses a `function*` syntax and `yield` keyword to pause and resume code execution halfway through.
