# The `return` Keyword in JavaScript

In JavaScript, the `return` keyword is used inside a function body to stop the function's execution and send a value back to the code that called it. [[1]](https://developer.mozilla.org/en-US/docs/Web/JavaScript/Reference/Statements/return) [[2]](https://www.geeksforgeeks.org/javascript/javascript-return-statement/)

## Syntax

```javascript
return [expression];
```

## Core Features

- **Stops Execution**: When JavaScript hits a `return` statement, it exits the function immediately. Any code written below it inside that block is unreachable.
- **Defaults to `undefined`**: If you omit the expression or use `return;` by itself, the function will explicitly output `undefined`.
- **Any Data Type**: You can return numbers, strings, arrays, objects, booleans, or even other functions. ## Code Examples

### 1. Returning a Value

Passes a computed result back to a variable outside the function. [[1]](https://www.youtube.com/watch?v=AGmba-D8H-M) [[2]](https://www.youtube.com/watch?v=4t8Q_YLpuKw&t=8)

```javascript
function add(a, b) {
  return a + b; // Calculates and sends back the sum
}

const result = add(5, 10);
console.log(result); // Output: 15
```

### 2. Early Exit (Guard Clauses)

Use `return` without a value to break out of a function early if a condition is met. [[1]](https://stackoverflow.com/questions/44474975/what-does-return-do-in-javascript) [[2]](https://www.youtube.com/watch?v=4t8Q_YLpuKw&t=8)

```javascript
function checkAge(age) {
  if (age < 18) {
    return; // Stops execution immediately if underage
  }
  console.log("Access granted."); // This will not run if age < 18
}
```

### 3. Returning Multiple Values (via Objects or Arrays)

A function can only execute one `return` statement, but you can package multiple pieces of data together. [[1]](https://www.naukri.com/code360/library/javascript-function-return-value) [[2]](https://www.youtube.com/watch?v=4t8Q_YLpuKw&t=8) [[3]](https://www.geeksforgeeks.org/javascript/javascript-return-statement/)

```javascript
function getUserData() {
  return {
    name: "Alex",
    role: "Developer"
  };
}
```

## Common Pitfalls

### Automatic Semicolon Insertion (ASI)

Never place your return value on a new line beneath the `return` keyword. JavaScript will automatically insert a semicolon after `return`, making the function output `undefined` and ignoring your code below it. [[1]](https://levelup.gitconnected.com/return-statements-in-javascript-c874b7ff8974) [[2]](https://medium.com/sessionstack-blog/how-javascript-works-the-different-ways-of-declaring-a-function-5-best-practices-8a0324c06fe2) [[3]](https://medium.com/@codeAbit/crazy-and-unexpected-behaviors-of-javascript-de10344b19a4) [[4]](https://ssojet.com/blog/20-funny-tricks-of-javascript-that-will-make-you-laugh-and-prob-cry)

**Incorrect:**

```javascript
return
{ name: "Alex" }; // Returns undefined!
```

**Correct:**

```javascript
return {
  name: "Alex"
};
```
