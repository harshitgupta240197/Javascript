# JavaScript Methods

In JavaScript, a **method** is a function that belongs to an object. While a standard function is executed on its own, a method is a property of an object containing a function definition and is invoked relative to that object.

## Custom Object Methods

You can define your own methods inside an object literal using either traditional syntax or the modern ES6 shorthand syntax.

```javascript
const user = {
  firstName: "Jane",
  lastName: "Doe",

  // ES6 Shorthand Method Syntax (Recommended)
  greet() {
    return `Hello, my name is ${this.firstName}.`;
  }
};

console.log(user.greet()); // Output: Hello, my name is Jane.
```

> **The `this` Keyword:** Inside an object method, the `this` keyword refers to the "owner" object (the object executing the code).

## Built-In JavaScript Methods

JavaScript includes robust built-in objects, each equipped with distinct data-manipulation methods.

### 1. Array Methods

Used to modify, loop through, or extract information from arrays.

- `.push()` / `.pop()` — Adds or removes items at the end of an array.
- `.map()` — Transforms every item in an array and returns a new array.
- `.filter()` — Returns a new array containing items that pass a specific condition.
- `.forEach()` — Loops through every item in an array without returning a new one.

### 2. String Methods

Used to manipulate and format text chunks. *(Note: Strings are immutable, so these methods return a new string.)*

- `.toUpperCase()` / `.toLowerCase()` — Changes the text casing.
- `.slice(start, end)` — Extracts a section of a string.
- `.replace(search, replace)` — Finds and replaces specific text within a string.
- `.trim()` — Removes whitespace from both ends of the string.

### 3. Object Methods

Static methods available globally via the `Object` constructor to manage structural properties.

- `Object.keys(obj)` — Returns an array of an object's keys.
- `Object.values(obj)` — Returns an array of an object's values.
- `Object.entries(obj)` — Returns an array of key-value pairs.

## Quick Comparison: Functions vs. Methods

| Feature | Function | Method |
| --- | --- | --- |
| **Association** | Declared standalone; independent. | Associated directly with an object or class. |
| **Invocation** | Called directly by name: `myFunc()`. | Called via an object reference: `obj.myMethod()`. |
| **Context (`this`)** | Context depends on execution environment. | Implicitly bound to the owner object. |
