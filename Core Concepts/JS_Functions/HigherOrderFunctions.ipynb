# Higher-Order Functions in JavaScript

A **higher-order function** is a function that either:

- accepts another function as an argument,
- returns a function as its output, or
- does both.

This is possible because JavaScript treats functions as **first-class citizens** — they can be assigned to variables, passed around, and manipulated just like numbers or strings.

---

## Built-in Array Higher-Order Functions

The most common built-in higher-order functions are array methods that process data without manual loops.

### `map()`

Transforms every element in an array and returns a **new array of the same length**.

```javascript
const numbers = [1, 2, 3];
const doubled = numbers.map(num => num * 2); // [2, 4, 6]
```

### `filter()`

Evaluates each element against a condition and returns a **new array containing only the items that match**.

```javascript
const scores = [45, 80, 95, 60];
const passing = scores.filter(score => score >= 70); // [80, 95]
```

### `reduce()`

Iterates through elements to accumulate them into a **single value** — a sum, an object, or anything else.

```javascript
const expenses = [10, 20, 30];
const total = expenses.reduce((sum, current) => sum + current, 0); // 60
```

## Custom Higher-Order Functions

You can write your own higher-order functions to abstract operations or build custom logic factories.

### 1. Accepting a function as an argument

When you pass a function into a higher-order function, the passed function is called a **callback function**.

```javascript
function repeat(n, action) {
  for (let i = 0; i < n; i++) {
    action(i);
  }
}

// Passing an anonymous function as the callback
repeat(3, (index) => console.log(`Iteration ${index}`));
// Logs: Iteration 0, Iteration 1, Iteration 2
```

### 2. Returning a function

A higher-order function can act as a **factory**, creating specific behaviours while leveraging **closures** to remember surrounding variables.

```javascript
function createMultiplier(factor) {
  return function (number) {
    return number * factor;
  };
}

const double = createMultiplier(2);
const triple = createMultiplier(3);

console.log(double(5)); // 10
console.log(triple(5)); // 15
```

## Why Use Higher-Order Functions?

| Benefit | What it gives you |
| --- | --- |
| **Code reusability** | Abstract recurring structural patterns (looping, timing) away from the specific logic you want to execute. |
| **Declarative code** | Focus on *what* you want to accomplish rather than writing step-by-step instructions on *how* to do it. |
| **Functional composition** | Combine simple, single-purpose functions to manage complex data pipelines. |
