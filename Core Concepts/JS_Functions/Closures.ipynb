# Closures in JavaScript

## Definition

A **closure** is the combination of a function bundled together with references to its surrounding state, known as the **lexical environment**. In simpler terms, a closure gives an inner function access to the outer function's scope even after the outer function has finished executing.

Every time a function is created in JavaScript, a closure is automatically created at function creation time — you don't have to do anything special to "make" one.

## How Closures Work

JavaScript uses **lexical scoping**, meaning the variable scope of a function is determined by its physical location within the source code. When functions are nested, the inner function retains a reference to the variables declared in its parent's scope.

Normally, local variables inside a function are cleaned up from memory via garbage collection once the function returns. However, if an inner function is still accessible (e.g., it was returned or passed as a callback), JavaScript keeps those outer variables alive in memory so the inner function can continue using them.

> **Key insight:** The inner function doesn't get a *copy* of the outer variable — it keeps a live *reference* to it. If the variable changes, the closure sees the updated value.

## Code Example: A Private Counter

```javascript
function createCounter() {
  let count = 0; // Local variable in outer scope

  return function() {
    count++; // Inner function accesses and modifies outer variable
    return count;
  };
}

const counter = createCounter();

console.log(counter()); // Output: 1
console.log(counter()); // Output: 2
console.log(counter()); // Output: 3
```

In the example above, `createCounter` finishes running, but the returned inner function forms a closure over the `count` variable. The variable `count` is now completely private and can only be manipulated by calling `counter()`.

### A subtle trap: closures inside loops

This is one of the most commonly asked closure interview questions, so it's worth having in the notes even though it wasn't in the original.

```javascript
// Problem: using 'var'
for (var i = 1; i <= 3; i++) {
  setTimeout(() => console.log(i), 1000);
}
// Output: 4, 4, 4  (not 1, 2, 3!)

// Fix: using 'let'
for (let i = 1; i <= 3; i++) {
  setTimeout(() => console.log(i), 1000);
}
// Output: 1, 2, 3
```

`var` is function-scoped, so all three callbacks close over the *same* `i`, which has already reached 4 by the time the timeouts fire. `let` is block-scoped, so each loop iteration gets its own fresh `i`, and each closure captures a distinct value.

## Common Use Cases

- **Data Encapsulation** — creating private variables and methods that cannot be accessed directly from the global scope (as in the counter example above).
- **State Preservation** — retaining state across asynchronous operations, such as event listeners, `setTimeout` callbacks, or API requests.
- **Function Factories** — creating reusable functions that share the same body logic but maintain distinct, separate lexical environments.
- **Currying** — transforming a function that takes multiple arguments into a sequence of nested functions that each take one argument at a time.

```javascript
// Currying example
function multiply(a) {
  return function(b) {
    return a * b;
  };
}

const double = multiply(2);
console.log(double(5)); // Output: 10
```

- **Memoization** — caching expensive computation results in a closed-over variable so repeated calls with the same input skip the recalculation.

```javascript
function memoize(fn) {
  const cache = {};
  return function(n) {
    if (n in cache) return cache[n];
    const result = fn(n);
    cache[n] = result;
    return result;
  };
}
```

## Real-World Application: React's `useState`

Closures are heavily used under the hood in popular UI libraries. React's functional components rely on closures to manage state via the `useState` hook. The hook returns a state getter and setter that close over an internal memory cell, ensuring your component always interacts with the correct, isolated state instance across re-renders.

## Quick Recap

| Concept | One-line summary |
|---|---|
| Closure | A function + the lexical scope it was created in |
| Lexical scoping | Scope determined by where code is *written*, not where it's *called* |
| Garbage collection | Skipped for variables still referenced by an active closure |
| Common uses | Encapsulation, state preservation, factories, currying, memoization |
