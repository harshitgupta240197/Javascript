# Function Scope in JavaScript

Function scope in JavaScript means that variables declared inside a function are completely isolated and cannot be accessed from outside that function. Each new function creates its own unique scope boundary.

## Key Characteristics

- **Enclosed Accessibility**: Variables defined inside a function are local to that function.
- **Outer Scope Isolation**: Code running outside the function will throw a `ReferenceError` if it tries to read or modify internal variables.
- **Keywords**: Variables declared with `var`, `let`, and `const` all respect function scope boundaries.
- **Nested Access**: Inner nested functions can look upward and access variables from their parent functions via lexical scoping.

## Code Example

```javascript
function processPayment() {
    var orderTotal = 150; // Function scoped
    let discountCode = "SAVE10"; // Function scoped
    const taxRate = 0.08; // Function scoped

    console.log(orderTotal); // Works fine inside the function
}

processPayment();

// Throws ReferenceError: orderTotal is not defined
console.log(orderTotal);
```

## Function Scope (`var`) vs. Block Scope (`let` / `const`)

While all three keywords are locked inside a function, they behave differently inside standard code blocks (like `if` statements or `for` loops):

| Feature         | `var`                                      | `let` & `const`                                      |
|-----------------|--------------------------------------------|------------------------------------------------------|
| Scope Boundary  | Scoped strictly to the nearest function.   | Scoped to the nearest curly braces `{}` (block).     |
| Block Leaking   | Escapes out of `if` blocks and loops.      | Trapped inside `if` blocks and loops.                |

### Example of "Leaking" Behavior

```javascript
function scopeCheck() {
    if (true) {
        var functionScoped = "I leak outside this IF block!";
        let blockScoped = "I am trapped inside this IF block!";
    }

    console.log(functionScoped); // Logs successfully because var ignores the IF block.
    console.log(blockScoped); // Throws ReferenceError.
}
```
