# Function Expressions in JavaScript

In JavaScript, a **function expression** is a syntax construct that allows you to define a function directly inside an expression, typically by assigning it to a variable.

Unlike traditional function declarations, function expressions treat the function as a normal data value (like a string or a number). This makes them highly flexible tools for modern programming.

## Basic Syntax

A standard function expression omits the function name (making it an *anonymous function*) and assigns it to a variable name.

```javascript
// Function Expression
const calcAge = function (birthYear) {
  return 2026 - birthYear;
};

// Invoking the function using the variable name
console.log(calcAge(10)); // Output: 2016
```

## Function Expression vs. Function Declaration

The main operational difference between a function expression and a function declaration is **hoisting**.

| Feature        | Function Declaration                        | Function Expression                                          |
|----------------|---------------------------------------------|--------------------------------------------------------------|
| Syntax         | `function name() {}`                        | `const name = function() {}`                                 |
| Hoisting       | Yes. Can be called before its definition.   | No. Cannot be accessed before initialization.                |
| End Semicolon  | No semicolon required after closing brace.  | Semicolon is required (it terminates an assignment statement). |

### Hoisting Code Example

```javascript
// 1. THIS WORKS (Function Declaration is hoisted)
greetDeclaration();

function greetDeclaration() {
  console.log("Hello from Declaration!");
}

// 2. THIS CRASHES (Function Expression is NOT hoisted)
greetExpression(); // ReferenceError: Cannot access 'greetExpression' before initialization

const greetExpression = function() {
  console.log("Hello from Expression!");
};
```

## Variants of Function Expressions

### 1. Named Function Expressions

You can optionally include a name inside the function expression. This name is only accessible inside the function's own body, making it useful for tracking errors in call stacks or managing recursive calls.

```javascript
const factorial = function compute(n) {
  if (n <= 1) return 1;
  return n * compute(n - 1); // "compute" is valid here
};

// compute(5); // ReferenceError: compute is not defined (Invalid outside)
```

### 2. Arrow Function Expressions

Introduced in ES6, arrow functions provide a much more concise syntax for writing function expressions.

```javascript
// Concise single-line expression with implicit return
const multiply = (a, b) => a * b;
```

## Practical Use Cases

Function expressions are incredibly useful because they exist exactly where they are evaluated. This makes them ideal for:

**Callbacks** — Passing a function directly into another function as an argument (such as array methods like `.map()` or event listeners).

```javascript
const numbers = [1, 2, 3];
const doubled = numbers.map(function(num) { return num * 2; });
```

**Immediately Invoked Function Expressions (IIFE)** — Defining a function and executing it instantly to encapsulate code and preserve scope.

```javascript
(function() {
  console.log("This runs right away without polluting global scope!");
})();
```
