# Returning a Function from a Function in JavaScript

Returning a function from another function is possible because JavaScript functions are **first-class citizens** — they can be treated like any other value (strings, numbers, objects). That means they can be:

- assigned to variables,
- passed as arguments, and
- returned from other functions.

---

## Basic Syntax

When a function returns another function, you call the outer function first. That returns the inner function, which you can then execute later.

```javascript
function createMultiplier(multiplier) {
  // Returns an anonymous function expression
  return function (number) {
    return number * multiplier;
  };
}

// 1. Call the outer function to get the inner function
const double = createMultiplier(2);

// 2. Call the inner function
console.log(double(5));  // Output: 10
console.log(double(10)); // Output: 20
```

> **Note:** You can also invoke the returned function immediately using double parentheses:
> ```javascript
> createMultiplier(2)(5); // Output: 10
> ```

## Core Concepts & Use Cases

### 1. Closures (state retention)

When an inner function is returned, it keeps access to the variables in its outer scope — **even after the outer function has finished executing**. This behaviour is called a **closure**, and it's widely used to create private variables.

```javascript
function createCounter() {
  let count = 0; // Private variable

  return function () {
    count++;
    return count;
  };
}

const counter = createCounter();
console.log(counter()); // Output: 1
console.log(counter()); // Output: 2
// `count` cannot be accessed or modified from the outside directly
```

### 2. Function factories (customised functions)

Write a generic outer function that generates specific variants of an inner function based on the parameters passed in.

```javascript
function greet(greeting) {
  return function (name) {
    return `${greeting}, ${name}!`;
  };
}

const sayHello = greet("Hello");
const sayHola = greet("Hola");

console.log(sayHello("Alice")); // Output: Hello, Alice!
console.log(sayHola("Bob"));    // Output: Hola, Bob!
```

### 3. Currying

**Currying** is a functional-programming technique where a function taking multiple arguments is transformed into a sequence of nested functions, each taking a single argument.

```javascript
// Regular function
const addRegular = (a, b) => a + b;

// Curried function using arrow shorthand
const addCurried = a => b => a + b;

console.log(addRegular(2, 3)); // Output: 5
console.log(addCurried(2)(3)); // Output: 5
```

## Common Pitfall: Automatic Semicolon Insertion (ASI)

When returning a function (or any value), **never put the returned value on a new line directly below `return`**. JavaScript inserts a semicolon after the bare `return`, and the function outputs `undefined`.

**❌ Incorrect**

```javascript
function getFunc() {
  return  // <-- JavaScript inserts a semicolon here!
  function () { console.log("Hi"); };
}
```

**✅ Correct**

```javascript
function getFunc() {
  return function () {
    console.log("Hi");
  };
}
```
