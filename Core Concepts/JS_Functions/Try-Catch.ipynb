# `try...catch` in JavaScript

In JavaScript, `try...catch` is a statement used to handle runtime errors without stopping or crashing the entire application. It lets you test a block of code for errors and provides a recovery path if something goes wrong.

## Basic Syntax

```javascript
try {
  // Code that might throw an error
} catch (error) {
  // Code that runs if an error occurs
}
```

- **`try` block** — Contains the fragile code you want to test.
- **`catch` block** — Executes only if an error is thrown inside the `try` block. The `error` parameter provides an object containing error details (like `error.name` and `error.message`).

## Core Components

### 1. The Standard `try...catch`

If an error happens in the `try` block, execution immediately jumps to the `catch` block.

```javascript
try {
  let user = JSON.parse('{"age": 30}'); // Valid JSON
  console.log(user.name); // undefined (not an error)

  let badUser = JSON.parse("{invalid json}"); // Throws a SyntaxError
  console.log("This line will never run");
} catch (error) {
  console.error("Error caught: " + error.message);
  // Output: Error caught: Unexpected token i in JSON at position 1
}
```

### 2. Optional Catch Binding

If you do not need to inspect the specific error details, you can omit the `(error)` parameter completely.

```javascript
try {
  performAction();
} catch {
  console.log("An error occurred, but I don't care about the details.");
}
```

### 3. The `finally` Block

You can append an optional `finally` block. It always executes, regardless of whether the code succeeded or failed. It is typically used for cleanup tasks like closing network connections or resetting loading spinners.

```javascript
try {
  openFile();
  processData();
} catch (err) {
  console.error("Failed to process data.");
} finally {
  closeFile(); // Always runs to prevent memory leaks
}
```

### 4. The `throw` Statement

You can intentionally create and fire your own errors using `throw`. This immediately stops the `try` block and moves execution directly into the `catch` block.

```javascript
try {
  let balance = -5;
  if (balance < 0) {
    throw new Error("Balance cannot be negative!"); // Custom error
  }
} catch (error) {
  console.error(error.message); // Output: Balance cannot be negative!
}
```

## Asynchronous Limitations (`async` / `await`)

A standard `try...catch` cannot catch errors in asynchronous code (like `setTimeout` or raw Promises) because the `try` block finishes executing before the asynchronous callback runs.

```javascript
// ❌ THIS WILL CRASH THE APP (Not caught)
try {
  setTimeout(() => {
    throw new Error("Delayed crash");
  }, 1000);
} catch (e) {
  console.log("Caught?"); // Will never run
}
```

**The Solution:** Use `async`/`await`. Wrapping the `await` keyword in a `try...catch` block successfully intercepts asynchronous errors.

```javascript
// ✅ THIS WORKS PERFECTLY
async function fetchData() {
  try {
    let response = await fetch("https://example.com");
    let data = await response.json();
    return data;
  } catch (error) {
    console.error("Network request failed:", error.message);
  }
}
```
