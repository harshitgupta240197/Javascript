# Lexical Scope in JavaScript

Lexical scope (also known as **static scope**) means that the accessibility of variables is determined strictly by the physical location of the code where those variables are declared. JavaScript parses variable availability at compile time, meaning a function's scope is locked in when you write the code, not when you execute it. [[1]](https://www.geeksforgeeks.org/javascript/why-is-lexical-scoping-important/) [[2]](https://www.geeksforgeeks.org/javascript/lexical-scope-in-javascript/)

## The Core Rule: Upward Availability Only

In JavaScript, an inner function always has full access to the variables defined in its outer (parent) function or block. However, a parent function cannot look downward into a child function to access its variables. [[1]](https://medium.com/@catherineisonline/what-is-lexical-scope-in-javascript-708b1d1487ea) [[2]](https://rahul319sinha.medium.com/understanding-scope-lexical-environment-and-the-scope-chain-in-javascript-c219cbb73417) [[3]](https://www.geeksforgeeks.org/javascript/why-is-lexical-scoping-important/)

## Code Example

```javascript
function outerFunction() {
  const outerVar = "I am outside!";

  function innerFunction() {
    // Looks upward and finds outerVar via lexical scoping
    console.log(outerVar);
  }

  innerFunction();
}

outerFunction(); // Logs: "I am outside!"
```

## How JavaScript Resolves Variables: The Scope Chain

When your code references a variable, the JavaScript engine looks for it using a bottom-up process called the **Scope Chain**: [[1]](https://www.educative.io/answers/lexical-scope-in-javascript) [[2]](https://rahul319sinha.medium.com/understanding-scope-lexical-environment-and-the-scope-chain-in-javascript-c219cbb73417) [[3]](https://dev.to/catherineisonline/what-is-lexical-scope-in-javascript-4gi8) [[4]](https://w3.cs.jmu.edu/mayfiecs/cs343/notes/wk07/cheat/js-cheatsheet.pdf) [[5]](https://medium.com/@shubh9911am/javascript-closures-most-asked-javascript-interview-question-4cf2e502fba8)

1. **Local Scope**: It checks inside the currently executing function or block.
2. **Outer Scope**: If not found locally, it steps one level up into the parent environment.
3. **Global Scope**: It continues moving up the hierarchy until it hits the global scope.
4. **Error**: If the variable isn't found anywhere along this chain, it throws a `ReferenceError`. ## Lexical Scope vs. Dynamic Scope

- **Lexical Scope (JavaScript)**: Scope is determined by *where the function is written*.
- **Dynamic Scope (Other languages)**: Scope is determined by *where the function is called*.
