# Async / Await in JavaScript

Asynchronous programming in JavaScript allows the application to remain responsive by executing long-running tasks — network requests, database operations, timers — in the background. Because JavaScript is single-threaded, synchronous execution blocks the main thread, freezing user interfaces during heavy tasks.

Modern JavaScript resolves this using the `async` and `await` keywords, which act as syntactic sugar built on top of native Promises to make asynchronous code look and feel synchronous.

---

## 🔑 Core Mechanics

### `async` functions

Preceding a function definition with the `async` keyword forces it to **always return a Promise**. If the function explicitly returns a non-promise value, JavaScript automatically wraps it in a resolved Promise.

### `await` keyword

Valid exclusively inside an `async` function (or at the top level of modern JavaScript modules). It pauses function execution until the targeted Promise resolves or rejects, directly extracting the resolved value.

---

## 💻 Practical Implementation

### Standard network fetch pattern

Using `async` / `await` completely eliminates `.then()` and `.catch()` callback chaining.

```javascript
// Declaring the asynchronous wrapper
async function getUserData(userId) {
  try {
    // Execution pauses here until the network request resolves
    const response = await fetch(`https://example.com/api/users/${userId}`);

    if (!response.ok) {
      throw new Error('Network response was not ok');
    }

    // Execution pauses here until JSON body conversion completes
    const data = await response.json();
    return data;

  } catch (error) {
    // Errors (including network failures) are caught natively here
    console.error("Failed to fetch user:", error.message);
  }
}

// Invoking the async function
getUserData(123).then(user => console.log(user));
```

---

### Parallel vs. sequential `await`

When multiple **independent** operations exist, placing `await` on consecutive lines causes them to run sequentially — each one blocks the next from starting. Grouping them concurrently improves performance.

| Execution type | Code pattern | Behavioural performance |
| --- | --- | --- |
| **Sequential (slow)** | `const dataA = await fetchA();`<br>`const dataB = await fetchB();` | `fetchB` does not start until `fetchA` completes. Total time is additive. |
| **Concurrent (fast)** | `const [dataA, dataB] = await Promise.all([fetchA(), fetchB()]);` | Both requests trigger simultaneously. Total time is bound only by the slowest single request. |

> **Note:** only use `Promise.all` when the operations are genuinely independent. If `fetchB` needs a value produced by `fetchA`, sequential `await` is correct, not slow.
