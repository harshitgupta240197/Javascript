# Callback Hell

**Callback Hell** (also known as the **Pyramid of Doom**) is a common anti-pattern in JavaScript where multiple nested callback functions are chained together to handle dependent asynchronous operations. The code grows horizontally rather than vertically, making it heavily indented, difficult to read, error-prone, and hard to maintain.

## The Core Problem

When several asynchronous tasks must happen in a specific order — e.g. fetching a user, getting their posts, then downloading comments — early JavaScript conventions forced developers to pass callbacks inside callbacks.

```javascript
// The Pyramid of Doom
getUser(userId, (user) => {
    getPosts(user.id, (posts) => {
        getComments(posts[0].id, (comments) => {
            renderDashboard(user, posts, comments, () => {
                console.log("Dashboard rendered!");
                // More nested logic...
            });
        });
    });
});
```

## Why It's Harmful

- **Poor readability** — the code drifts off the right side of the screen, making the actual execution flow hard to follow.
- **Brittle error handling** — errors must be handled manually at every single level, leading to duplication or unhandled bugs.
- **Inversion of control** — by passing your function to a third-party API, you blindly trust it to invoke your callback correctly, at the right time, and only once.

## How to Escape It

Modern JavaScript introduced two primary solutions to flatten async code.

### 1. Promises (ES6+)

Promises flatten the structure through chaining with `.then()` and `.catch()`.

```javascript
getUser(userId)
    .then(user => getPosts(user.id))
    .then(posts => getComments(posts[0].id))
    .then(comments => renderDashboard(comments))
    .catch(error => console.error("An error occurred:", error)); // Single error handler
```

### 2. Async/Await (ES2017+)

Built on top of Promises, `async/await` is the most modern approach. It lets you write asynchronous code that reads sequentially, like synchronous code.

```javascript
async function setupDashboard(userId) {
    try {
        const user = await getUser(userId);
        const posts = await getPosts(user.id);
        const comments = await getComments(posts[0].id);
        await renderDashboard(comments);
        console.log("Dashboard rendered!");
    } catch (error) {
        console.error("An error occurred:", error); // Handled via standard try/catch
    }
}
```
