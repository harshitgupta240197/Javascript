# JavaScript Promises

A **Promise** is an object representing the eventual completion or failure of an asynchronous operation. It acts as a placeholder for a value that isn't known yet when the promise is created, allowing asynchronous methods to return values much like synchronous ones do.

## The Three States

A promise always exists in exactly one of three mutually exclusive states:

- **Pending** — initial state; the asynchronous operation is still running.
- **Fulfilled** — the operation completed successfully, returning a result value.
- **Rejected** — the operation failed, returning an error or reason for failure.

Once a promise transitions to fulfilled or rejected, it is **settled** and its state becomes permanent.

## Creating a Promise

Instantiate with the `new Promise()` constructor. It takes an *executor* function with two built-in callback arguments: `resolve` and `reject`.

```javascript
const fetchUserData = new Promise((resolve, reject) => {
  const success = true; // Simulating an operation outcome
  setTimeout(() => {
    if (success) {
      resolve({ id: 1, name: "Alex" }); // Moves state to Fulfilled
    } else {
      reject("Failed to fetch user data."); // Moves state to Rejected
    }
  }, 1000);
});
```

## Consuming a Promise

Attach consumer methods to handle the outcome:

- `.then()` — runs if the promise fulfills; receives the resolved value.
- `.catch()` — runs if the promise is rejected; handles the error.
- `.finally()` — runs regardless of the final state, ideal for cleanup like stopping a loading spinner.

```javascript
fetchUserData
  .then((user) => {
    console.log("User found:", user.name);
  })
  .catch((error) => {
    console.error("Error:", error);
  })
  .finally(() => {
    console.log("Operation completed.");
  });
```

## Promise Chaining

Every `.then()` or `.catch()` call returns a **new** promise. This lets you sequence multiple asynchronous steps without nesting into a deep [callback hell](callback-hell.md) structure.

```javascript
fetchUserData
  .then((user) => {
    // Return a new value or promise to pass down the chain
    return user.id;
  })
  .then((userId) => {
    console.log("Processing ID:", userId);
  })
  .catch((err) => {
    console.error("Caught error from any step above:", err);
  });
```

## Advanced Combinators (Static Methods)

For handling multiple promises concurrently:

| Method | Behavior | Use Case |
|---|---|---|
| `Promise.all([...])` | Resolves when **all** succeed; rejects instantly if any single one fails. | Batching independent network requests together. |
| `Promise.allSettled([...])` | Waits until all inputs finish, returning an array of their individual status and values. | Tracking outcomes when you don't care if some failed. |
| `Promise.race([...])` | Settles as soon as the **first** promise finishes — success or failure. | Implementing custom timeouts for slow operations. |
| `Promise.any([...])` | Resolves as soon as the first **successful** promise fulfills. | Fetching from redundant mirror servers where one success suffices. |

## Modern Alternative: Async / Await

`async` and `await` are syntactic sugar built directly on top of promises, making asynchronous code look and behave more like synchronous code.

```javascript
async function getDashboardData() {
  try {
    const user = await fetchUserData; // Pauses execution until promise resolves
    console.log("Loaded dashboard for:", user.name);
  } catch (error) {
    console.error("Dashboard error:", error); // Catches promise rejections
  }
}
```
