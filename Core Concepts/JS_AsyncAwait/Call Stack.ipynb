# The JavaScript Call Stack

The **Call Stack** is a mechanism used by the runtime engine to track function execution in a script. It acts as a synchronous "to-do list" that tells the engine which function is currently running and which functions are called from within it.

Because JavaScript is **single-threaded**, it can only execute one task at a time. The Call Stack manages this single thread using the **Last-In, First-Out (LIFO)** data structure principle.

## How the Call Stack Works

- **Global Context** — When a script loads, the engine creates a Global Execution Context and pushes it to the bottom of the stack.
- **Push** — Every time you invoke a function, a new stack frame containing its arguments and local variables is pushed onto the top of the stack.
- **Execute** — The engine prioritizes the function currently at the very top of the stack.
- **Pop** — When a function finishes or encounters a `return` statement, its frame is popped off the top, and execution resumes where it left off in the function below it.

## Step-by-Step Code Example

Consider this nested execution flow:

```javascript
function greet() {
    console.log("Hello!");
}

function processUser() {
    greet();
}

processUser();
```

The Call Stack mutates state-by-state during execution like this:

| State | Call Stack (Top → Bottom) | Action Description |
|-------|---------------------------|--------------------|
| 1 | `[ Global Context ]` | Script begins executing. |
| 2 | `[ processUser ]` `[ Global Context ]` | `processUser()` is invoked and pushed to the top. |
| 3 | `[ greet ]` `[ processUser ]` `[ Global Context ]` | Inside `processUser`, `greet()` is called and pushed to the top. |
| 4 | `[ console.log ]` `[ greet ]` `[ processUser ]` `[ Global Context ]` | Inside `greet`, `console.log()` is pushed to the top. |
| 5 | `[ greet ]` `[ processUser ]` `[ Global Context ]` | `"Hello!"` prints. `console.log` finishes and is popped off. |
| 6 | `[ processUser ]` `[ Global Context ]` | `greet()` reaches its end and is popped off the stack. |
| 7 | `[ Global Context ]` | `processUser()` finishes and is popped off the stack. |
| 8 | `[ ]` (empty stack) | The script completes execution. |

## What is a Stack Overflow?

A **stack overflow** occurs when the call stack exceeds its hard-allocated memory limit. This is almost always caused by **infinite recursion** — a function calling itself repeatedly without a proper base case to break the loop.

```javascript
// This function will crash the browser/runtime engine
function recurse() {
    recurse();
}

recurse(); // RangeError: Maximum call stack size exceeded
```

Because `recurse()` never hits a `return` statement, frames are continuously pushed onto the stack without ever being popped, causing the runtime to throw a `RangeError`.

## Interaction with Asynchronous JavaScript

Because the Call Stack is synchronous, long-running synchronous code blocking the stack can freeze a web page. To manage asynchronous actions (like `setTimeout`, `fetch` requests, or event listeners), JavaScript uses an **Event Loop** alongside a **Callback Queue**.

Asynchronous callbacks wait in a separate queue and are only pushed onto the Call Stack by the Event Loop **after the stack becomes entirely empty**.
