# JavaScript Object Literals

#### Objects are collections of properties --> Properties are a key-value pair --> Rather than accessing data using an index, we use custom keys

#### All keys are converted to string

An object literal in JavaScript is a comma-separated list of key-value pairs wrapped in curly braces (`{}`), providing the simplest and most common way to create an object. It allows you to define a structured data type instantly without needing a class blueprint or constructor function.

## Core Syntax and Anatomy

```javascript
const user = {
  // Key: Value
  firstName: "Jane",          // String property
  age: 28,                    // Number property
  "current status": "Active", // Multi-word keys must be quoted
  isAdmin: true,              // Boolean property
};
```

## Accessing Object Properties

You can read or update properties using two different methods:

1. **Dot Notation:** `user.firstName` (Best for standard, single-word keys)
2. **Bracket Notation:** `user["current status"]` (Mandatory for keys with spaces, special characters, or dynamic variables)

## Advanced Object Literal Features

Modern JavaScript (ES6 and later) introduces powerful shorthands to make writing object literals more concise:

### 1. Property Shorthand

If a variable name matches the property key name, you can omit the colon and the value.

```javascript
const speed = 120;

// Instead of { speed: speed }, write:
const car = { speed }; 
```

### 2. Method Shorthand

You can drop the `function` keyword and colon when assigning functions (methods) inside an object.

```javascript
const calculator = {
  // Shorthand syntax
  add(a, b) {
    return a + b;
  }
};
```

### 3. Computed Property Names

You can dynamically calculate a property key at runtime by placing an expression inside square brackets.

```javascript
const keyName = "score";
const status = {
  [keyName]: 95 // Evaluates to score: 95
};
```

### 4. The Spread Operator (`...`)

You can use the spread operator to instantly copy or merge object properties into a new object literal.

```javascript
const baseConfig = { theme: "dark", notifications: true };
const userConfig = { ...baseConfig, notifications: false, language: "en" };
// Result: { theme: "dark", notifications: false, language: "en" }
```

## Key Behavioral Rules

- **Duplicate Keys:** If you declare the same key twice in an object literal, the final value will overwrite the previous ones.
- **Reference Types:** Assigning an object literal to a new variable copies its reference, not its values. Mutating the new variable alters the original object.
- **Const Declarations:** Creating an object using `const` prevents you from reassigning the variable to a new object, but you can freely mutate, add, or delete properties within that object.

# Modifying Objects in JavaScript

In JavaScript, you can modify an object mutably by directly updating its properties, or immutably by creating a new copy with the changes. The most direct approach to modify an existing object is using dot notation (`obj.key = value`) or bracket notation (`obj['key'] = value`) to update or add properties.

## Direct Mutation (Modifying the Original Object)

### 1. Dot Notation

Best for static, known property names.

```javascript
const user = { name: 'Alice', age: 25 };
user.age = 26; // Updates existing property
user.country = 'USA'; // Adds new property
```

### 2. Bracket Notation

Required if the property name contains spaces, special characters, or is stored in a dynamic variable.

```javascript
const user = { name: 'Alice' };
const keyName = 'user-age';

user[keyName] = 25; // Dynamic key assignment
user['home address'] = '123 Main St'; // Space in key name
```

### 3. Deleting Properties

Use the `delete` operator to entirely remove a property from an object.

```javascript
const user = { name: 'Alice', age: 25 };
delete user.age; // Removes 'age' property
```

## Immutable Modification (Creating a New Object)

In modern frameworks like React, mutating the original object directly is discouraged. Instead, use these methods to create a modified clone.

### 1. Spread Operator (`...`)

The cleanest way to merge or update properties while preserving the original object (shallow copy).

```javascript
const original = { name: 'Alice', age: 25 };
const updated = { ...original, age: 26, role: 'Admin' }; 
// original remains unchanged; updated has new values
```

### 2. `Object.assign()`

Copies properties from source objects to a target object.

```javascript
const target = { a: 1 };
const source = { b: 2, c: 3 };
const result = Object.assign({}, target, source); // Merges into a fresh object
```

## Advanced Restrictions

If you want to control how or if an object can be modified, JavaScript provides built-in static helper methods:

- **`Object.freeze(obj)`**: Completely locks the object. You cannot add, delete, or modify any existing properties.
- **`Object.seal(obj)`**: Prevents adding or deleting properties, but allows you to modify the values of existing properties.
- **`Object.defineProperty(obj, prop, descriptor)`**: Allows defining a property with specific permissions (e.g., setting `writable: false` to make a single property read-only).

# Nested Arrays and Objects in JavaScript

In JavaScript, nested arrays and objects are combined structures used to represent complex, real-world data models like API responses.

## The Core Difference

- **Objects** store data as key-value pairs (`{ key: value }`) and are accessed using dot notation (`obj.key`) or bracket notation (`obj["key"]`).
- **Arrays** store ordered lists of data (`[item1, item2]`) and are accessed using zero-based numerical indices via bracket notation (`arr[0]`).

## Anatomy of a Nested Structure

Below is a typical real-world data structure combining both nested objects and nested arrays:

```javascript
const userProfile = {
  id: 101,
  name: "Sarah Jenkins",
  contact: {
    email: "sarah@example.com",
    phone: {
      home: "555-0123",
      work: "555-9876"
    }
  },
  preferences: ["dark-mode", "notifications-on"],
  orders: [
    { id: "A1", total: 45.99, items: ["book", "pen"] },
    { id: "A2", total: 12.50, items: ["notebook"] }
  ]
};
```
