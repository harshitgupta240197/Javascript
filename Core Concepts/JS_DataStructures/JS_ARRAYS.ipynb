# JavaScript Arrays

In JavaScript, an array is an ordered, globally-accessible object used to store multiple values of mixed data types under a single variable name. JavaScript arrays are zero-indexed (the first element is at index `0`) and have dynamic sizes that automatically adjust when items are added or removed.

## Basic Operations

You can create, read, modify, and track arrays using simple, standard syntax:

```javascript
// Creation via literal syntax (recommended)
const fruits = ['Apple', 'Banana', 'Cherry'];

// Accessing elements
console.log(fruits[0]); // Output: 'Apple'

// Modifying elements
fruits[1] = 'Blueberry';

// Checking length
console.log(fruits.length); // Output: 3
```
--------
## Essential Array Methods

JavaScript provides built-in methods to manipulate data structures cleanly. You can find comprehensive documentation on these on the [MDN Web Docs Array Page](https://developer.mozilla.org/en-US/docs/Web/JavaScript/Reference/Global_Objects/Array).

### Adding & Removing Elements

- `push()`: Appends elements to the end of the array.
- `pop()`: Removes and returns the final element.
- `unshift()`: Adds elements to the beginning of the array.
- `shift()`: Removes and returns the first element.

### Iterating & Searching

- `forEach()`: Executes a provided callback function once for each element.
- `indexOf()`: Returns the first index where a specific element is found.
- `includes()`: Determines if an array contains a certain value, returning true/false.

### Transformation & Filtering

These methods return a new array without changing the original one:

- `map()`: Creates a new array populated with the results of calling a function on every element.
- `filter()`: Creates a new array with all elements that pass a test condition.
- `slice()`: Extracts a section of an array and returns it as a new array object.

# JavaScript: push, pop, shift, unshift

In JavaScript, `push`, `pop`, `shift`, and `unshift` are built-in methods used to add or remove elements from an array. They directly modify (mutate) the original array.

## Overview Table

| Method | Action | Location | Returns | Performance (Time Complexity) |
|---|---|---|---|---|
| `push()` | Adds element(s) | End of array | New array length | O(1) (Fast) |
| `pop()` | Removes element | End of array | The removed element | O(1) (Fast) |
| `unshift()` | Adds element(s) | Beginning of array | New array length | O(n) (Slower — reindexes array) |
| `shift()` | Removes element | Beginning of array | The removed element | O(n) (Slower — reindexes array) |

## 1. Operations at the End of the Array

### `push(...items)` and `pop()`

- **`push(...items)`**: Appends one or more elements to the end of an array, returning the updated new length.
- **`pop()`**: Extracts and returns the final element from the array. Returns `undefined` if the array is empty.

## 2. Operations at the Beginning of the Array

### `unshift(...items)` and `shift()`

- **`unshift(...items)`**: Inserts new elements at the start of the array and outputs the new length.
- **`shift()`**: Eliminates the first element, returning it while reindexing remaining elements down by one position. Returns `undefined` for empty arrays.

# JavaScript: concat, indexOf, includes, reverse

`concat`, `indexOf`, `includes`, and `reverse` are core JavaScript array methods used to merge, search, verify, and reorder data structures. The primary differences in their behavior, parameters, and structural impact are outlined in the comparison table below.

## Method Overview Matrix

| Method | Purpose | Mutation Status | Returns |
|---|---|---|---|
| `concat()` | Joins two or more arrays or values. | Immutable (Creates a new copy). | A fresh merged array. |
| `indexOf()` | Finds the first index position of a value. | Immutable (Reads data only). | The target index (Number) or `-1`. |
| `includes()` | Checks if an element exists inside the array. | Immutable (Reads data only). | A boolean (`true` or `false`). |
| `reverse()` | Inverts the order of elements. | Mutable (Changes the source array). | The mutated reference. |

## Detailed Breakdowns & Implementations

### 1. `Array.prototype.concat()`

The `concat()` method merges arrays or single values without modifying the original inputs. It automatically unwraps first-level array arguments during combination.

**Use Case:** Creating safe backups or assembling distributed data chunks.

```javascript
const frameworkGroupA = ['React', 'Vue'];
const frameworkGroupB = ['Angular', 'Svelte'];

// Merge arrays and individual items seamlessly
const combinedFrameworks = frameworkGroupA.concat(frameworkGroupB, 'SolidJS');

console.log(combinedFrameworks);
// Output: ['React', 'Vue', 'Angular', 'Svelte', 'SolidJS']

console.log(frameworkGroupA);
// Output: ['React', 'Vue'] (Untouched)
```

### 2. `Array.prototype.indexOf()`

The `indexOf()` method searches the array from left to right. It uses strict equality (`===`) to match your search target and halts processing immediately upon finding the first match.

**Syntax Variation:** `array.indexOf(searchElement, fromIndex)` allows specifying a custom search start location.

**Limitation:** It fails to identify `NaN` values due to internal JavaScript strict comparison rules.

```javascript
const activeUsers = ['Alice', 'Bob', 'Charlie', 'Bob'];

console.log(activeUsers.indexOf('Bob'));       // Output: 1 (First match location)
console.log(activeUsers.indexOf('Dominick'));  // Output: -1 (Not found)
console.log(activeUsers.indexOf('Bob', 2));    // Output: 3 (Search began past index 2)
```

### 3. `Array.prototype.includes()`

The `includes()` method delivers clean boolean results. Unlike `indexOf()`, it correctly detects `NaN` elements because it uses the modern SameValueZero algorithm internally.

**Use Case:** Writing readable, conditional `if` statements.

```javascript
const premiumInventory = ['Laptop', 'Tablet', NaN];

if (premiumInventory.includes('Laptop')) {
  console.log('Product catalog verified.'); // Code executes
}

console.log(premiumInventory.includes('Phone')); // Output: false
console.log(premiumInventory.includes(NaN));     // Output: true (indexOf would return -1)
```

### 4. `Array.prototype.reverse()`

The `reverse()` method physically shuffles array slots directly in memory. If you must preserve original datasets, use the modern `toReversed()` alternative or duplicate the structure beforehand using a spread operator (`[...]`).

**Risk Factor:** Any changes impact all variables pointing to that original reference link.

```javascript
const systemQueue = [10, 20, 30];

// In-place structural transformation
systemQueue.reverse();

console.log(systemQueue);
// Output: [30, 20, 10] (Original structure updated)
```

# JavaScript: slice, splice, sort

In JavaScript, `slice()` extracts a portion of an array into a new array without changing the original, while `splice()` directly modifies the original array by adding, removing, or replacing elements, and `sort()` rearranges the original array's elements in place.

## Quick Comparison Table

| Method | Mutates Original Array? | Primary Purpose | Returns |
|---|---|---|---|
| `slice()` | ❌ No | Copies or extracts a subset | A brand new array |
| `splice()` | ✅ Yes | Deletes, inserts, or replaces items | Array of deleted elements |
| `sort()` | ✅ Yes | Reorders the elements | The same array (reordered) |

## 1. The `slice()` Method

`slice()` is a non-mutating method. It copies a piece of your array.

**Syntax:** `array.slice(startIndex, endIndex)`

**Key Detail:** The `endIndex` is exclusive (not included in the result).

**Omitted Arguments:** If you leave out arguments, it clones the entire array.

```javascript
const fruits = ['Apple', 'Banana', 'Orange', 'Mango', 'Kiwi'];

// Extract elements from index 1 up to (but not including) index 4
const subArray = fruits.slice(1, 4);

console.log(subArray); // ['Banana', 'Orange', 'Mango']
console.log(fruits);   // ['Apple', 'Banana', 'Orange', 'Mango', 'Kiwi'] (Unchanged)
```

## 2. The `splice()` Method

`splice()` is a destructive, mutating method. Use it to add, change, or remove items.

**Syntax:** `array.splice(startIndex, deleteCount, item1, item2, ...)`

**Key Detail:** It returns what was removed from the array.

```javascript
const months = ['Jan', 'March', 'April', 'June'];

// At index 1, delete 0 elements, and insert 'Feb'
months.splice(1, 0, 'Feb');
console.log(months); // ['Jan', 'Feb', 'March', 'April', 'June']

// At index 3, delete 1 element ('April'), and insert 'May'
const deleted = months.splice(3, 1, 'May');
console.log(months);  // ['Jan', 'Feb', 'March', 'May', 'June']
console.log(deleted); // ['April'] (The removed item)
```

## 3. The `sort()` Method

`sort()` sorts elements in place.

**Warning:** By default, `sort()` converts elements to strings and compares their UTF-16 code values. This causes unexpected results with numbers (e.g., `10` comes before `2`).

**Fix:** Always provide a compare function when sorting numbers.

```javascript
const numbers = [40, 100, 1, 5, 25];

// Incorrect default behavior:
numbers.sort();
console.log(numbers); // [1, 100, 25, 40, 5] ❌

// Correct numeric sort (Ascending):
numbers.sort((a, b) => a - b);
console.log(numbers); // [1, 5, 25, 40, 100]

// Correct numeric sort (Descending):
numbers.sort((a, b) => b - a);
console.log(numbers); // [100, 40, 25, 5, 1]
```

# `const` Arrays in JavaScript

When you declare an array with `const` in JavaScript, it does **not** make the array elements immutable — it only makes the array *reference* constant. You cannot reassign the variable to a new array, but you can freely modify, add, or remove the elements inside it.

## Quick Summary

| Action | Allowed? | Example |
|---|---|---|
| Modify elements | ✅ Yes | `myArray[0] = "New Value";` |
| Add/Remove elements | ✅ Yes | `myArray.push("Item");` |
| Reassign the variable | ❌ No | `myArray = ["New", "Array"];` (Throws Error) |

## What You Can Do (Mutable Values)

Because `const` only protects the reference to the array's memory address, the following operations work perfectly:

```javascript
const numbers = [1, 2, 3];

// 1. Change an element value
numbers[0] = 99; 

// 2. Add an element
numbers.push(4); 

console.log(numbers); // Output: [99, 2, 3, 4]
```

## What You Cannot Do (Immutable Reference)

You will trigger a `TypeError: Assignment to constant variable` if you try to reassign the identity of the array:

```javascript
const numbers = [1, 2, 3];

// ❌ Throws an error: You cannot point 'numbers' to a brand new array
numbers = [4, 5, 6]; 

// ❌ Throws an error: Even trying to clear it completely fails
numbers = []; 
```

## How to Make a `const` Array Truly Unchangeable

If you want to prevent anyone from modifying the elements inside the array, pass it to `Object.freeze()`:

```javascript
const staticList = Object.freeze(["Apple", "Banana"]);

// Modifying or adding elements will now fail silently (or throw an error in strict mode)
staticList[0] = "Orange"; 
staticList.push("Grape"); 

console.log(staticList); // Output: ["Apple", "Banana"]
```

# JavaScript Multidimensional Arrays

In JavaScript, there is no built-in syntax for "true" multidimensional arrays; instead, you create them by nesting arrays within other arrays (often called jagged arrays).

Here is a quick overview of how to create, access, and manipulate a 2D array.

## 1. Creation and Initialization

### Using Array Literals

The easiest way to initialize a grid or matrix structure:

```javascript
const matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
];
```

### Dynamically Creating an Empty Grid

To generate a grid of a specific size (e.g., a 3 × 3 grid filled with zeros) dynamically:

```javascript
const rows = 3;
const cols = 3;
const grid = Array.from({ length: rows }, () => Array(cols).fill(0));
```

## 2. Accessing and Modifying Elements

You can pinpoint elements using chained square brackets `[row][column]`. Remember that JavaScript utilizes zero-based indexing.

- **Read a value:** `matrix[1][2]` returns `6` (second row, third column).
- **Write/Modify a value:** `matrix[0][0] = 99` changes the top-left element to `99`.

## 3. Iterating Over Elements

To traverse every element in a 2D array, use nested loops.

```javascript
// Using modern for...of loops
for (const row of matrix) {
    for (const element of row) {
        console.log(element);
    }
}
```

## 4. Common Built-in Methods

Because the sub-elements are normal arrays, you can use regular JavaScript array methods on them:

- **Add a row:** `matrix.push([10, 11, 12]);`
- **Add an item to a specific row:** `matrix[0].push(4);`
- **Flatten the array:** `matrix.flat();` turns a 2D array into a simple 1D array (`[1, 2, 3, 4, 5...]`).
